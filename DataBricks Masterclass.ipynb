# 📘 Delta Lake — An Experienced Data Engineer’s Lecture

---

## 1️⃣ Opening the Lecture (Context Setting)

> *“Data lakes gave us cheap storage and scalability, but they never gave us reliability.  
Delta Lake was created to fix that exact gap.”*

As data engineers, we store massive volumes of data in **object storage** such as:
- S3
- ADLS
- GCS

But traditional data lakes come with serious challenges:
- No transactions
- No consistency
- No easy updates
- No reliable concurrent access

**Delta Lake exists to bring reliability to data lakes** 

---

## 2️⃣ What Delta Lake Is (And Is Not)

### ✅ Delta Lake *IS*:
- An **open-source storage framework**
- A **storage layer on top of data lakes**
- A key enabler of **Lakehouse architecture**

### ❌ Delta Lake is *NOT*:
- Proprietary technology
- A storage medium
- A data warehouse or database service

> **Think of Delta Lake as a reliability layer, not a new storage system**

---

## 3️⃣ Why Delta Lake Was Needed (Real Engineering Problems)

Before Delta Lake, we used **Parquet files directly** in object storage.

Problems we faced:
- Partial writes corrupt data
- Concurrent writes overwrite each other
- No rollback if a job fails
- Schema changes break downstream jobs
- Metadata does not scale

The key missing piece was:
> **A transaction system for object storage**

That missing piece is **Delta Lake**.

---

## 4️⃣ Delta Lake Core Architecture (Simple & Fundamental)

At its core, a Delta table consists of:



![Delta Table Architecture](Delta_table.png)


### Key insight:
> **Parquet files store data, Delta log stores truth**

---

## 5️⃣ The Delta Transaction Log (_delta_log)

The **transaction log** is the heart of Delta Lake.

### What the Delta log contains:
- Ordered records of every transaction
- Operation performed (INSERT, UPDATE, DELETE, MERGE)
- Files added or removed
- Predicates used

> The Delta log is the **single source of truth** for a table 

Each commit is stored as a **JSON file**.

---

## 6️⃣ How Reads and Writes Work in Delta Lake

### ✍️ Write Operation

1. Writer creates new Parquet files
2. Writer writes a new entry to `_delta_log`
3. Commit is atomic

### 📖 Read Operation

1. Reader reads the transaction log
2. Determines which files are valid
3. Reads only those data files

```text
Writer → Parquet Files → Delta Log
Reader → Delta Log → Parquet Files
```
This ensures consistency and correctness
________________________________________

## 7️⃣ How Updates Work (Copy-on-Write)

Delta Lake uses **copy-on-write**.

### Example:
- Update affects **File 2**
- Delta creates **File 3** (updated copy)
- Delta log marks **File 2** as removed
- **File 3** becomes active

Old files are **not overwritten**, which enables:
- Time travel  
- Rollbacks  
- Auditing  

👉 This ensures **consistency and correctness**.

---

## 8️⃣ Concurrent Reads and Writes (ACID in Action)

Delta Lake supports **simultaneous readers and writers**.

### How it works:
- Readers always read from a **consistent snapshot**
- Writers create **new versions**
- Delta log **serializes commits**

### This enables:
- Streaming + batch on the same table  
- Multiple pipelines writing safely  

> **ACID transactions on object storage** is the real breakthrough.

---

## 9️⃣ What Happens During Failed Writes?

If a job fails mid-write:
- Parquet file **may exist**
- **BUT** no Delta log commit is written

### Result:
- Readers ignore that file
- Data remains consistent

👉 This is how Delta Lake prevents **corrupt tables**.

---

## 🔟 ACID Transactions Explained Simply

Delta Lake guarantees:

- **Atomicity** – All or nothing  
- **Consistency** – Valid states only  
- **Isolation** – Readers never see partial writes  
- **Durability** – Committed data persists  

All of this works **without a traditional database**.

---

## 1️⃣1️⃣ Schema Enforcement & Evolution

Delta Lake protects your pipelines by:
- Rejecting invalid schema writes  
- Allowing controlled schema evolution  

👉 This prevents **silent data corruption**, a common data lake problem.

---

## 1️⃣2️⃣ Time Travel (One of the Most Powerful Features)

Delta Lake keeps **historical versions** of data.

You can:
- Query older versions  
- Roll back bad data  
- Audit changes  

This is possible because files are **immutable and versioned**.

---

## 1️⃣3️⃣ Delta Lake Advantages (From Real Projects)

Delta Lake provides:
- ACID transactions on object storage  
- Scalable metadata handling  
- Full audit trail of all changes  
- Built on open formats: **Parquet + JSON**

This makes Delta Lake:
- Reliable  
- Scalable  
- Cloud-agnostic  

---

## 1️⃣4️⃣ Delta Lake in Lakehouse Architecture

Delta Lake enables:
- Data lake flexibility

👉 This is why the **Databricks Lakehouse** is built on Delta Lake.

## 1️⃣5️⃣ When to Use Delta Lake

## 3️⃣ Querying Older Versions of Data

Delta Lake allows querying previous snapshots of a table.

⏳ Using Timestamp
```sql
SELECT * 
FROM my_table 
TIMESTAMP AS OF '2019-01-01';
```
🔢 Using Version Number
```sql
SELECT * 
FROM my_table 
VERSION AS OF 36;
```
Shorthand
```sql
SELECT * FROM my_table@v36;
```
Key Insight:
👉 You are not querying files — you are querying logical table versions.
________________________________________

## 4️⃣ Rolling Back Data (RESTORE TABLE)

Time travel is not only for querying — it can fix production issues.

Roll back using timestamp
```sql 
RESTORE TABLE my_table 
TO TIMESTAMP AS OF '2019-01-01';
```
Roll back using version
```sql
RESTORE TABLE my_table 
TO VERSION AS OF 36;
```
What happens internally:

- No files are overwritten

- Delta log creates a new version

- Older files become active again

Key Insight:
👉 RESTORE is metadata-driven, not file rewriting.
________________________________________
## 5️⃣ Compacting Small Files (OPTIMIZE)
The Small File Problem

In streaming or frequent batch jobs:

- Many small Parquet files are created

- Query performance degrades

- Metadata overhead increases

Delta Lake solves this using compaction.

🧹 OPTIMIZE Command
```sql
OPTIMIZE my_table;
```

### What it does:

- Combines many small files

- Creates fewer, larger files

- Improves query performance

### Key Insight:
👉 OPTIMIZE improves read performance, not write speed.
________________________________________

## 6️⃣ Indexing with Z-ORDER (Data Skipping Optimization)
Delta Lake does not use traditional indexes.
Instead, it uses data skipping.

### Z-ORDER Concept

- Co-locates related column values

- Improves predicate filtering

- Reduces data scanned

### 📌 Z-ORDER Example
```sql
OPTIMIZE my_table
ZORDER BY (id);
```

### Best use cases:

- High-cardinality columns

- Frequently filtered columns

- Join keys

### Key Insight:
👉 Z-ORDER is most effective after OPTIMIZE, not before.
________________________________________
## 7️⃣ Vacuum — Cleaning Up Old Data Files
### Delta Lake keeps old files for:
- Time travel

- Rollbacks

- Auditing

But storage grows continuously.
### VACUUM cleans up:

- Uncommitted files

- Files no longer referenced by the latest table state

### 🗑️ VACUUM Command
```sql
VACUUM my_table;
```
### Default retention period:

- 7 days

### Custom retention
```sql
VACUUM my_table RETAIN 168 HOURS;
```

### ⚠️ Important Warning

VACUUM = No Time Travel beyond retention period

Once files are vacuumed:

- Old versions cannot be queried

- RESTORE is not possible beyond retention window

________________________________________
## 8️⃣ Production Best Practices (Key Insights)
### ✅ Time Travel

- Essential for debugging

- Use DESCRIBE HISTORY regularly

### ✅ OPTIMIZE

- Schedule during low-load windows

- Especially important for streaming tables

### ✅ Z-ORDER

- Apply only on frequently queried columns

- Avoid overuse (costly operation)

### ✅ VACUUM

- Never reduce retention casually

- Coordinate with compliance and audit requirements

### Golden Rule:
👉 Performance, reliability, and auditability must be balanced.

# 📘 Optimizing Data File Layout in Delta Lake
________________________________________
## 🎯 Learning Objectives
By the end of this session, you will be able to:
- Understand what data file layout means in Delta Lake
-	Explain why file layout optimization is critical for performance
-	Apply different optimization techniques based on query patterns
-	Choose the right strategy between Partitioning, Z-Ordering, and Liquid Clustering

________________________________________
## 📂 Understanding Data File Layout
Data file layout refers to how data is physically organized and stored inside a Delta table.
Although users interact with tables using SQL, under the hood the data is stored as multiple Parquet files managed by the Delta transaction log.
The way these files are arranged directly impacts:
- Query performance
- Amount of data scanned
- Cost of execution

Optimized layouts allow Databricks to skip unnecessary files during query execution, significantly reducing I/O.
________________________________________

## ⚡ Why Optimizing File Layout Matters
When a query runs, Databricks:
1.	Reads table metadata
2.	Identifies relevant files
3.	Skips files that cannot match query conditions
4.	Scans only required data

Efficient layout means fewer files scanned, which results in:
- Faster queries
- Lower compute usage
- Better scalability for large datasets


________________________________________
## 🛠 Optimization Techniques in Delta Lake
Delta Lake provides multiple techniques to optimize data layout:
- Partitioning
- Z-Order Indexing
- Liquid Clustering

Each technique serves a different use case and must be chosen carefully based on data characteristics and query patterns.


________________________________________
## 📁 Partitioning
Partitioning organizes a table by grouping rows that share the same value for one or more columns.
Each partition is stored as a separate directory.
### Example Table Data
| id | name  | month | year |
| -- | ----- | ----- | ---- |
| 1  | Adam  | 9     | 2015 |
| 2  | Sarah | 11    | 2015 |
| 3  | John  | 5     | 2000 |
| 4  | Ali   | 9     | 2000 |
| 5  | Emma  | 10    | 2000 |
| 6  | Eric  | 1     | 2023 |

### Creating a Partitioned Table
```sql
CREATE TABLE my_table (
  id INT,
  name STRING,
  year INT,
  month INT
)
USING DELTA
PARTITIONED BY (year);
```
This creates separate directories such as:
```text
year=2000/
year=2015/
year=2023/
```
________________________________________
## 🚀 Partition Skipping
When a query filters on the partition column, Databricks automatically skips unrelated partitions.
```sql
SELECT *
FROM my_table
WHERE year = 2023;
```
Only the year=2023 directory is scanned, improving performance significantly.
________________________________________
## ⚠ Limitations of Partitioning
While partitioning is powerful, it has important drawbacks:
- File compaction cannot happen across partitions
- Leads to a small files problem
-	Inefficient for high-cardinality columns
- Repartitioning requires a full table rewrite
-	Partitioning strategy is static
Because of these limitations, partitioning must be used cautiously.
________________________________________
## 🧩 Z-Order Indexing
Z-Order indexing reorganizes data inside files to colocate related values together without creating directories.

It improves performance by enhancing file-level data skipping, especially for high-cardinality columns.

### Applying Z-Order
```sql
OPTIMIZE my_table
ZORDER BY (id);
```
After Z-Ordering, records with similar id values are stored closer together within the same files.
________________________________________
## ⚠ Z-Order Behavior
Z-Ordering is not incremental.
Each time the OPTIMIZE ... ZORDER command is executed:
- Existing data may be reshuffled
-	New data is not automatically merged with old layout
-	Repeated optimizations can be expensive

This makes Z-Order less suitable for frequently updated tables.
________________________________________
## 🌊 Liquid Clustering
Liquid Clustering is a modern optimization technique that improves upon Z-Ordering with greater flexibility and performance.
### Key characteristics:
- Defined at the table level
-	Incremental by design
-	No directory-based partitions
-	Better suited for evolving workloads

Creating a Clustered Table
```sql
CREATE TABLE table1 (
  col1 INT,
  col2 STRING,
  col3 DATE
)
USING DELTA
CLUSTER BY (col1, col3);
```
Adding Clustering to an Existing Table
```sql
ALTER TABLE table2
CLUSTER BY (col1);
```
Liquid Clustering cannot be used together with Partitioning or Z-Ordering.
________________________________________
## 🔁 Incremental Clustering Behavior
With Liquid Clustering:

- Existing optimized data remains untouched
-	New incoming data is incrementally clustered
-	No full table rewrite is required

This makes it ideal for streaming and continuously growing datasets.
________________________________________
## 🔑 Choosing Clustering Keys
Clustering keys should be selected based on how the table is queried.
### Best practices:
-	Choose columns frequently used in WHERE, JOIN, or GROUP BY
-	Avoid extremely high-cardinality combinations
-	Start with 1–2 columns and adjust over time

One major advantage is flexibility — clustering keys can be redefined without rewriting existing data.
________________________________________
### 🤖 Automatic Liquid Clustering
Databricks can automatically choose clustering keys by analyzing historical query patterns.
This feature:

-	Requires Unity Catalog managed tables
-	Uses Predictive Optimization
-	Is ideal when query workloads are unpredictable

Enabling Automatic Clustering
```sql
CREATE TABLE logs (...)
USING DELTA
CLUSTER BY AUTO;
ALTER TABLE logs
CLUSTER BY AUTO;
```
Databricks continuously adapts the layout based on actual usage.
________________________________________
### 🧠 Summary
-	File layout optimization is critical for scalable analytics
-	Partitioning works best for low-cardinality, static columns
-	Z-Ordering improves skipping for high-cardinality columns but is not incremental
-	Liquid Clustering offers flexible, incremental, and modern optimization
-	Automatic clustering removes manual tuning for dynamic workloads

Choosing the right strategy leads to faster queries, lower costs, and better system efficiency.

## 🗂 Databases in Databricks

In Databricks, databases are logical containers used to organize tables and views.
Technically, Databricks databases are equivalent to schemas in the Hive Metastore.

You can create a database using either syntax:
```sql
CREATE DATABASE db_name;
```

or
```sql
CREATE SCHEMA db_name;
```

Both commands are functionally identical.

## 🧠 Hive Metastore

The Hive Metastore acts as a central metadata repository.

It stores information such as:

- Databases (schemas)

- Tables

- Table locations

- Column definitions

- Table properties

The metastore does not store actual data, only metadata about where the data lives and how it is structured.

## 🗄 Default Storage Behavior

When no custom location is specified:

- Databricks stores databases and tables under a default warehouse path

- This path is typically:
```text
dbfs:/user/hive/warehouse
```

### Example
```sql
CREATE TABLE table1;
CREATE TABLE table2;
```

### Result:

- Table data is stored inside the default warehouse directory

- Metadata is registered in the Hive Metastore

# 📁 Database-Specific Storage

When a new database is created, Databricks creates a corresponding directory inside the warehouse.
```sql
CREATE SCHEMA db_x;
USE db_x;

CREATE TABLE table1;
CREATE TABLE table2;
```

### Storage structure:

- Each database has its own folder

- Tables created inside the database are stored within that folder

This helps logically and physically separate data across databases.

## 📍 Using the LOCATION Keyword with Databases

Databricks allows you to control where a database’s data is stored using the LOCATION keyword.
```sql
CREATE SCHEMA db_y
LOCATION 'dbfs:/custom/path/db_y.db';
```

### After this:

- Metadata is still stored in the Hive Metastore

- Actual table data is stored in the specified custom path

### This is useful when:

- You want data stored outside the default warehouse

- You want better control over storage organization

### 📊 Tables in Databricks

Databricks supports two main types of tables:

- Managed tables

- External tables

Understanding the difference is critical for data lifecycle management.

## 📦 Managed Tables

Managed tables are fully controlled by Databricks.

### Characteristics:

- Created inside the database directory

- No LOCATION keyword specified

- Databricks manages both metadata and data
```sql
CREATE TABLE table_name;
``` 

### Behavior on drop:

- Dropping the table deletes the underlying data files

- Metadata and storage are both removed

Managed tables are ideal when Databricks owns the complete data lifecycle.

## 🔗 External Tables

External tables store data outside the database directory.

### Characteristics:

- Created using the LOCATION keyword

- Metadata is managed by Databricks

- Data is managed externally
```sql
CREATE TABLE table_name
LOCATION 'dbfs:/some/path/table_name';
```

### Behavior on drop:

- Dropping the table removes only metadata

- Underlying data files remain intact

### External tables are useful when:

- Data is shared across systems

- Data lifecycle is managed externally

## 📂 External Tables Across Databases

The same storage location can be referenced from different databases using different table definitions.

### Example:
```sql
USE db_x;
CREATE TABLE table3
LOCATION 'dbfs:/some/path_2/x_table3';

USE db_y;
CREATE TABLE table3
LOCATION 'dbfs:/some/path_2/y_table3';

```
### Each database:

- Registers its own metadata entry

- Can point to different external storage locations

This allows flexibility in multi-database environments.

## 🔍 Impact of the LOCATION Keyword

The LOCATION keyword directly affects:

- Where data is stored

- Whether a table is managed or external

- What happens to data when a table is dropped

| Scenario               | Metadata | Data     |
| ---------------------- | -------- | -------- |
| Managed table dropped  | Deleted  | Deleted  |
| External table dropped | Deleted  | Retained |

## 🧠 Key Takeaways

Databases in Databricks map to schemas in the Hive Metastore

The Hive Metastore stores metadata, not actual data

Default storage is under the Databricks warehouse directory

The LOCATION keyword gives control over physical storage

Managed tables delete data on drop

External tables preserve data on drop
