# 📘 AWS Lambda – Complete Lecture Note for Data Engineering

---

## 🔷 1. Introduction to AWS Lambda

AWS Lambda is a **serverless compute service** that allows you to run code **without managing servers**.

It is primarily used for:
- Event-driven processing
- Automation
- Lightweight data transformations

---

## 🔷 2. Why AWS Lambda?

Traditional compute requires:
- Server provisioning
- Scaling management
- Continuous running cost

AWS Lambda solves this by:
- Running code only when triggered
- Automatically scaling
- Charging only per execution time

---

## 🔷 3. Core Features of AWS Lambda

### ✅ 3.1 Serverless Execution
- No server management
- Fully managed by AWS

---

### ✅ 3.2 Event-Driven Architecture
Lambda runs when triggered by events such as:
- File upload (S3)
- API request (API Gateway)
- Database update (DynamoDB)

---

### ✅ 3.3 Automatic Scaling
- Scales automatically with number of requests
- Handles thousands of concurrent executions

---

### ✅ 3.4 Multi-Language Support
Supports:
- Python
- Node.js
- Java
- Go
- .NET

---

### ✅ 3.5 Short Execution Duration
- Maximum execution time: **15 minutes**
- Ideal for quick processing tasks

---

### ✅ 3.6 Pay-per-Use Pricing
- Charged based on:
  - Execution time
  - Memory usage

---

### ✅ 3.7 Easy Integration with AWS Services
- Native integration with most AWS services

---

### ✅ 3.8 Stateless Execution
- Each execution is independent
- No memory of previous runs

---

## 🔷 8. Sample Lambda Function (Python)

```python
import boto3

def lambda_handler(event, context):
    glue = boto3.client('glue')

    response = glue.start_job_run(
        JobName='flight_data_glue_job'
    )

    return {
        'statusCode': 200,
        'body': 'Glue job triggered successfully'
    }
```

## 🔷 9. AWS Lambda vs AWS Glue (Quick Comparison)

| Feature | AWS Lambda | AWS Glue |
|--------|-----------|---------|
| Purpose | Event-driven compute | ETL processing |
| Data Size | Small/Medium | Large (Big Data) |
| Runtime | Max 15 mins | Long-running |
| Processing | Single function | Distributed (Spark) |

---

## 🔷 10. Limitations of AWS Lambda

- ❌ Maximum execution time (15 minutes)  
- ❌ Limited memory (~10 GB)  
- ❌ Not suitable for large joins  
- ❌ No distributed processing  

---

## 🔷 11. Best Practices

- Keep functions small and focused  
- Use Lambda for triggering, not heavy processing  
- Combine with Glue for ETL workloads  
- Handle errors and retries properly  

---

## 🔷 12. Key Takeaways

- AWS Lambda is best for **event-driven, lightweight processing**  
- It complements AWS Glue, not replaces it  
- Ideal for automation, triggers, and microservices  
- Plays a key role in modern serverless architectures  

---

## 🎯 Final Summary

AWS Lambda enables scalable, event-driven computing in modern data engineering architectures. It is widely used for triggering workflows, automating pipelines, and handling lightweight processing tasks, especially in combination with AWS Glue for large-scale ETL operations.
