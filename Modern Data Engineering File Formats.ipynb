# 🗂️ Modern Data Engineering: File Formats Using PySpark (AWS Glue / EMR)

File formats play a critical role in:

- ✅ Storage optimization  
- ✅ Query performance  
- ✅ Scalability  

Using **PySpark**, we typically:

1. Read raw data  
2. Transform it  
3. Store it in optimized formats  

---

## 🔷 2. PySpark Setup

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("FileFormatsExample").getOrCreate()
```

---

## 🔷 3. CSV Format

### ✅ Characteristics
- Row‑based format  
- No schema enforcement  
- Human‑readable  

### 🔹 PySpark – Read CSV

```python
df_csv = spark.read.option("header", True).option("inferSchema", True).csv("s3://bucket/raw/data.csv")

df_csv.show()
```

### 🔹 PySpark – Write CSV

```python
df_csv.write.mode("overwrite").csv("s3://bucket/output/csv/")
```

### ❗ Use Case
- Raw ingestion layer  
- External system data  

### ❗ Limitations
- Large file size  
- Slow query performance  

---

## 🔷 4. JSON Format

### ✅ Characteristics
- Semi‑structured  
- Supports nested data  

### 🔹 PySpark – Read JSON

```python
df_json = spark.read.option("multiline", True).json("s3://bucket/raw/data.json")

df_json.printSchema()
```

### 🔹 PySpark – Flatten JSON

```python
from pyspark.sql.functions import col

df_flat = df_json.select(
    col("user.id"),
    col("user.name"),
    col("transaction.amount")
)
```

### 🔹 PySpark – Write JSON

```python
df_json.write.mode("overwrite").json("s3://bucket/output/json/")
```

### ❗ Use Case
- API data  
- Event/streaming data  

---

---

## 🔷 6. ORC Format

### ✅ Characteristics
- Columnar format  
- Optimized for Hive  

### 🔹 Write ORC

```python
df_csv.write.mode("overwrite").orc("s3://bucket/processed/orc/")
```

### 🔹 Read ORC

```python
df_orc = spark.read.orc("s3://bucket/processed/orc/")
```

### ❗ Use Case
- EMR + Hive workloads  

---

## 🔷 7. Avro Format

### ✅ Characteristics
- Row‑based binary format  
- Supports schema evolution  

### 🔹 Read Avro

```python
df_avro = spark.read.format("avro").load("s3://bucket/stream/avro/")
```

### 🔹 Write Avro

```python
df_csv.write.format("avro").mode("overwrite").save("s3://bucket/output/avro/")
```

### ❗ Use Case
- Kafka / Kinesis streaming pipelines  

---

## 🔷 8. Partitioning Strategy

### 🔹 Example Path
```
s3://bucket/data/year=2026/month=03/day=27/
```

### 🔹 PySpark – Partition Write

```python
df_parquet.write.mode("overwrite").partitionBy("year", "month").parquet("s3://bucket/partitioned/")
```

### 🚀 Benefits
- Faster queries  
- Reduced data scan  

---

## 🔷 9. Compression Strategy

### 🔹 PySpark – Compression

```python
df_parquet.write.option("compression", "snappy").parquet("s3://bucket/compressed/")
```

### 🔹 Types
- Snappy (default, fast)  
- Gzip (higher compression)

## 🔷 10. Schema Evolution

## 🔷 12. End‑to‑End Pipeline Example

```python
# Step 1: Read raw CSV
df = spark.read.option("header", True).csv("s3://raw/data.csv")

# Step 2: Transform
df_clean = df.filter("status = 'CONFIRMED'")

# Step 3: Write as Parquet
df_clean.write.mode("overwrite").parquet("s3://processed/data/")
```

## 🔷 13. Best Practices

- Convert CSV/JSON → Parquet  
- Use partitioning for large datasets  
- Use compression to reduce cost  
- Avoid small files (use coalesce/repartition)  

---

## 🔷 14. Key Takeaways

- **CSV/JSON → Raw ingestion**  
- **Parquet/ORC → Optimized analytics**  
- **Avro → Streaming pipelines**  
- **PySpark → Core engine for transformation**
