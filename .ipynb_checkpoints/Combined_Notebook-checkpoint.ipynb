### 1. Lists (Mutable, Ordered, Indexed) 📝
Lists are versatile, ordered collections that support mutation and dynamic sizing.
#### 1.Filter & Map: 
To return the squares of all stock prices currently above $\$100$, use a list comprehension to combine filtering and transformation:

```python
prices = [95, 110, 80, 150, 100]
squared_high_prices = [p**2 for p in prices if p > 100]
# Result: [12100, 22500]
```

#### 2. In-place Sort & Reverse:
   To sort a list of cities alphabetically, then reverse the order without creating new lists, use the list.sort() and list.reverse() methods:

```python
cities = ["London", "Paris", "Tokyo", "Berlin"]
cities.sort()
cities.reverse()
# Result: ['Tokyo', 'Paris', 'London', 'Berlin']
```

#### 3. Remove All:
   To remove all occurrences of the status code 404 from a list, use a list comprehension to filter the list, then assign the result back to the original list using slicing ([:]) to modify it in-place:

```python
status_codes = [200, 404, 301, 404, 500]
status_codes[:] = [c for c in status_codes if c != 404]
# Result: [200, 301, 500]
```

#### 4. Flatten:
   To flatten a list representing a coordinate grid into a single 1D list, use a nested list comprehension where the loops follow the same order as standard nested loops:

```python
matrix = [[1, 2], [3, 4], [5, 6]]
flat_list = [item for sublist in matrix for item in sublist]
# Result: [1, 2, 3, 4, 5, 6]
```

#### 5. Check Sort:
   To check if a list of sensor readings is monotonically increasing (non-decreasing), use the all() function combined with a generator expression to compare adjacent elements by index:

```python
sensor_data = [10, 12, 12, 15]
result = all(sensor_data[i] <= sensor_data[i+1] for i in range(len(sensor_data) - 1))
# Result: True
```

#### 6. Rotate:
   To rotate a list of tasks one step to the right, use list slicing combined with in-place assignment ([:]):

```python
tasks = ['A', 'B', 'C', 'D']; k = 1
tasks[:] = tasks[-k:] + tasks[:-k]
# Result: ['D', 'A', 'B', 'C']
```

#### 7. Deep Copy:
   To create a true independent copy of a list containing nested lists (employee records), you must use copy.deepcopy() to ensure changes in the new list do not affect the nested mutable objects in the original:

```python
import copy
original = [['Alice', 1], ['Bob', 2]]
new_copy = copy.deepcopy(original)
new_copy[0][1] = 99 # This change is isolated
print( original[0][1]) #remains 1
print(new_copy)
```

#### 8. Find Index:
   To find the index of the first month starting with 'M', use a generator expression with enumerate() (to get both index and value) inside the next() function:

```python
months = ['Jan', 'Feb', 'Mar', 'Apr']
index = next((i for i, m in enumerate(months) if m.startswith('M')), -1)
# Result: 2
```

#### 9. Pairwise Sum:
    To calculate the daily profit (sales - cost) from two lists, use the zip() function to iterate over elements from both lists simultaneously in a list comprehension:

```python
sales = [100, 200]; costs = [30, 80]
profit = [s - c for s, c in zip(sales, costs)]
# Result: [70, 120]
```

### 📘 Tuples in Python (Immutable, Ordered, Indexed)
---

#### 🧩 1. Unpack a Tuple

```python
coord = (40.7, -74.0, 100, 20)
lat, lon, *rest = coord
print('lat:', lat, '| lon:', lon, '| rest:', rest)
```

#### 🪜 2. Access Nested Tuple Values

```python
conf = ('server', ('ip', 403), 'admin')
status = conf[1][1]
print('Status:', status)
```

#### 🧠 3. Convert List to Tuple of Lengths

```python
words = ['hello', 'world', 'ai']
lengths = tuple(len(s) for s in words)
print(lengths)
```

#### 🌀 4. Repeat Values in a Tuple

```python
zero_tuple = (0,) * 5
print(zero_tuple)
```

#### 🔍 5. Check for an Element

```python
log_levels = ('info', 'warn', 'error', 'debug')
exists = 'error' in log_levels
print('Exists:', exists)
```

#### 🧾 6. Convert Tuple to String

```python
bits = ('1', '0', '1', '1')
binary_str = ''.join(bits)
print(binary_str)
```

#### ⚙️ 7. Convert Tuple List to Dictionary

```python
settings = [('theme', 'dark'), ('notify', True)]
settings_dict = dict(settings)
print(settings_dict)
```

#### 📊 8. Sort List of Tuples by Item

```python
people = [('Alice', 30), ('Bob', 25), ('Charlie', 35)]
people.sort(key=lambda x: x[1])
print(people)
```

#### 🔗 9. Concatenate Two Tuples

```python
h = ('ID', 'Name')
d = (101, 'Alice')
record = h + d
print(record)
```

#### 🔢 10. Count Occurrences

```python
traj = ((0, 0), (1, 2), (0, 0), (3, 4))
count = traj.count((0, 0))
print('Count:', count)
```

### 3. Sets (Mutable, Unordered, Unique) 🔑
Sets are unordered collections of unique, immutable elements, optimized for membership testing.

#### Remove Duplicates
The most Pythonic way to remove duplicates from a list is by converting it to a set and then back to a list:

```python
logs = ['1.1.1.1', '2.2.2.2', '1.1.1.1']
unique_ips = list(set(logs))
unique_ips
```

#### Intersection
To find products purchased by both User A and User B, use the intersection operator (&):

```python
A = {'apple', 'banana'}
B = {'banana', 'orange'}
both_bought = A & B
both_bought
```

#### Union
To get all unique device IDs from two network types, use the union operator (|):

```python
wifi = {1, 2}
cell = {3, 2}
all_devices = wifi.union(cell)
all_devices
```

#### Difference
To find customers subscribed to the Newsletter but not active, use the difference operator (-):

```python
subs = {'Alice', 'Bob'}
active = {'Alice', 'Charlie'}
missing_activation = subs - active
missing_activation
```

#### Symmetric Difference
To find members unique to only one of two project teams, use the symmetric difference operator (^):

```python
team_a = {'X', 'Y'}
team_b = {'Y', 'Z'}
unique_to_one = team_a ^ team_b
unique_to_one
```

#### Superset/Subset
To check if required permissions are a subset of the user's permissions, use the less-than-or-equal operator (<=):

```python
user_perms = {'read', 'write'}
req_perms = {'read'}
is_allowed = req_perms <= user_perms
is_allowed
```

#### Add Element
To add a single new element to a set, use the set.add() method:

```python
protocols = {'HTTP', 'HTTPS'}
protocols.add('FTP')
protocols
```

#### Frozen Set
To create a hashable set of column names for use as a dictionary key, use frozenset():

```python
columns = ['ID', 'Name']
frozen_cols = frozenset(columns)
frozen_cols
```

#### Set Comprehension
To create a set of unique status codes (as integers) from a list of strings, use a set comprehension:

```python
logs = ['404', '200', '404']
unique_codes = {int(c) for c in logs}
unique_codes
```

#### Discard vs. Remove
To safely attempt to remove a user ID without raising an error if it's absent, use set.discard():

```python
active_users = {101, 102}
active_users.discard(999)
active_users
```
