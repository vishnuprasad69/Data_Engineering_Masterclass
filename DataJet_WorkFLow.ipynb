# 📘 DataJet Project – Architecture & Flow Documentation

## 1️⃣ Project Overview  
**DataJet** is a modular, configuration-driven data ingestion and processing framework designed to:  
- Load data into **Bronze, Silver, Gold** layers  
- Perform validations and data quality checks  
- Support **config-driven onboarding**  
- Provide monitoring and observability  
- Run across platforms (**Databricks & AWS**)  
- Follow the **Medallion Architecture**  

---

## 🏗 2️⃣ High-Level Architecture  

## Data Flow

Source Data  
⬇️  
**Bronze Layer** (Raw ingestion)  
⬇️  
**Silver Layer** (Cleansed & transformed)  
⬇️  
**Gold Layer** (Aggregated / Business-ready)  
⬇️  
**Analytics / ML / Reporting**

## 🥉 3️⃣ Bronze Layer  
📂 **Files:**  
- `ingestion/internal/bronze_ingestion.py`  
- `BronzeIngestion-Commercials.py`  

### 🎯 Purpose  
- Ingest **raw data**  
- Apply minimal transformations  
- Add metadata (batch_id, timestamp, etc.)  
- Write to Bronze Delta/Parquet  

### 🔎 Responsibilities  
- Schema enforcement  
- Metadata enrichment  
- Basic validation  
- Raw zone persistence  

---

## 🥈 4️⃣ Silver Layer  
📂 **Files:**  
- `ingestion/internal/silver_processor.py`  
- `ingestion/internal/ingestion_engine.py`  

### 🎯 Purpose  
- Cleanse and standardize raw data  
- Apply business rules  
- Remove duplicates  
- Perform DQ checks  

### 🔎 Responsibilities  
- Transformations  
- Schema standardization  
- Handle bad/rejected records  
- Integrate DQ checks  

---

## 🥇 5️⃣ Gold Layer  
📂 **Files:**  
- `ingestion/internal/gold_ingestion.py`  
- `common/utils/gold_ingestion_utils.py`  

### 🎯 Purpose  
- Apply business aggregations  
- Prepare analytics-ready datasets  
- Build KPI and reporting tables  

### 🔎 Responsibilities  
- Aggregations  
- KPI calculations  
- Denormalized/consumption-ready tables  

---

## ⚙️ 6️⃣ Core Ingestion Engine  
📂 **File:**  
- `ingestion/DataJetCoreIngestion.py`  

### 🎯 Role  
Acts as the **central orchestrator**:  
- Controls flow across layers  
- Validates data  
- Reads configurations  
- Manages runtime context  

Coordinates modules like:  
- `ingestion_engine.py`  
- `ingestion_validators.py`  
- Platform handlers

## 📋 8️⃣ Configuration-Driven Design  
📂 **Directories:**  
- `configuration/`  
- `smartautomation/config/`  

### Config Levels  
- **L0** – Source metadata  
- **L1** – Table-level configuration  
- **L2** – Automation & code generation  

### Examples  
- `FLIGHT_LEG.json`  
- `PNR_DETAILS.json`  
- `COUPON_INFORMATION.json`  

### Design Pattern  
- Zero hardcoding  
- Full metadata-driven ETL  
- Rapid onboarding

3. Platform handler initialized  
4. Bronze ingestion  
5. Silver processing + DQ

7. Metric collection
