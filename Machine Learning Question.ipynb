### **ML Lesson 1: BCG Gamma - "Precision/Recall Tradeoff" (Easy)**

#### **The Problem**
Explain the tradeoff between Precision and Recall. In what situations would you prioritize one over the other?

#### **The Logic**
* **Precision:** Out of all the positive predictions the model made, how many were actually correct? (True Positives / (True Positives + False Positives)).
* **Recall:** Out of all the actual positive cases in reality, how many did the model successfully find? (True Positives / (True Positives + False Negatives)).
* **The Tradeoff:** As you lower the threshold to catch more positive cases (increasing Recall), you inevitably flag more negative cases by mistake (decreasing Precision). 
* **Prioritization:** Optimize for **Recall** in medical diagnoses (it's better to false-alarm a healthy patient than miss a deadly disease). Optimize for **Precision** in spam filters (it's better to let a spam email through than to accidentally send a critical work email to the junk folder).

#### **Senior Data Engineer Perspective**
* **Imbalanced Datasets:** In production, maximizing recall often requires oversampling minority classes or undersampling majority classes. A Senior DE must build robust data pipelines that can perform stratified sampling on terabytes of data during the ETL phase so that Data Scientists don't crash their Jupyter notebooks trying to balance a massive dataset in memory.

### **ML Lesson 2: McKinsey - "Assumptions of Linear Regression" (Easy)**

#### **The Problem**
What are the core assumptions of a Linear Regression model?

#### **The Logic**
Remember the acronym **LINE**:
1. **Linearity:** The relationship between the independent variables and the dependent variable is linear.
2. **Independence:** The observations are independent of each other (no autocorrelation).
3. **Normality:** The residuals (errors) of the model are normally distributed.
4. **Equal Variance (Homoscedasticity):** The variance of the residuals is constant across all levels of the independent variables.

#### **Senior Data Engineer Perspective**
* **Data Quality Frameworks:** A Data Scientist's linear regression model will fail silently if the upstream data violates these assumptions (e.g., a bug introduces massive outliers, ruining normality). A Senior DE implements automated Data Quality checks (like Anomalo or Great Expectations) that monitor the distribution, skewness, and variance of incoming data streams, automatically halting the pipeline if the data shape drifts.

### **ML Lesson 3: Blackstone - "Real Estate Modelling" (Medium)**

#### **The Problem**
How would you build a machine learning model to predict real estate prices? What features would you use, and how would you evaluate the model?

#### **The Logic**
* **Task:** This is a supervised regression problem.
* **Features:** Square footage, number of bedrooms/bathrooms, zip code/neighborhood, age of the property, proximity to transit, and macroeconomic indicators (interest rates).
* **Model Selection:** Gradient Boosted Trees (XGBoost, LightGBM) or Random Forests handle tabular data and non-linear relationships well.
* **Evaluation:** Root Mean Squared Error (RMSE) or Mean Absolute Error (MAE) to measure how far off the predicted prices are from actual sales in dollars.

#### **Senior Data Engineer Perspective**
* **Geospatial Processing at Scale:** "Proximity to transit" requires geospatial joins (e.g., calculating the Haversine distance between a house's lat/long and the nearest subway station's lat/long). Relational databases are terrible at this. A Senior DE uses specialized tools like PostGIS or PySpark with Apache Sedona to efficiently execute geospatial joins across millions of properties.

### **ML Lesson 4: LinkedIn - "Connection Request Spam" (Medium)**

#### **The Problem**
How would you design a system to detect and prevent connection request spam on a professional networking site?

#### **The Logic**
* **Task:** Supervised binary classification (Spam vs. Not Spam) or Unsupervised Anomaly Detection.
* **Features:** Velocity of requests (requests per minute), acceptance rate of past requests, account age, profile completeness, and network overlap (do they share mutual connections?).
* **Action:** If the spam probability is high, trigger a CAPTCHA or temporarily shadowban the user.

#### **Senior Data Engineer Perspective**
* **Streaming vs. Batch Architecture:** Catching spam tomorrow is useless. This model must execute in real-time. A Senior DE architects a streaming pipeline where user actions are pushed to an Apache Kafka topic. A stream processing engine (like Apache Flink or Spark Streaming) calculates the rolling "velocity of requests" feature over a 5-minute sliding window, passes it to the ML model API, and triggers an immediate block if the threshold is breached.

### **ML Lesson 5: Robinhood - "Modeling User Churn" (Hard)**

#### **The Problem**
Design a machine learning model to predict user churn for a financial trading app. How do you define churn, what features matter, and how do you handle the time-series nature of the data?

#### **The Logic**
* **Defining Churn:** Because trading apps don't usually have rigid monthly subscriptions, churn must be defined behaviorally (e.g., "0 logins and $0 balance for 30 consecutive days").
* **Features:** Frequency of logins, trailing 7-day/30-day deposit/withdrawal ratios, customer support tickets opened, and app crash telemetry.
* **Model:** Survival Analysis (predicting *when* someone will churn) or a time-windowed classification model (predicting *if* they will churn in the next 14 days).

#### **Senior Data Engineer Perspective**
* **The Feature Store:** Defining churn behaviorally means looking back in time. Creating rolling aggregates (like "30-day withdrawal ratio") for millions of users daily is computationally massive. A Senior DE implements a **Feature Store** (like Feast or Databricks Feature Store). The data engineering pipeline computes these complex aggregations once overnight, stores them in the Feature Store, and serves them at low latency to the ML model during both training and real-time inference.
