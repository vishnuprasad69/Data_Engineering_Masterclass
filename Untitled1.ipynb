# Prompt: Design Cloud Architecture for a Travel Analytics System

You are a principal cloud architect. Design a production-ready cloud architecture for a **Travel Analytics System**.

## **Context**
- The platform ingests clickstream, search, booking, and partner feed data.
- It powers near real-time analytics, dashboards, and ML-based insights.
- It must support both consumer‑facing APIs and B2B partner APIs.
- The system experiences global traffic with seasonal peak loads.

---

## **Deliverables (Use these as H2 sections in your answer)**

### **1. Cloud Provider Choice**
- Select ONE primary provider: AWS / Azure / GCP.
- Provide clear justification (analytics capabilities, networking, global availability, cost, enterprise maturity).
- Optionally provide equivalents for other providers.

### **2. Multi-AZ Deployment**
- Describe how the architecture spans multiple Availability Zones.
- Identify services that are zonal vs regional.
- Include failover strategy inside the primary region.

### **3. VPC / VNet & Network Isolation**
- Use a Hub–Spoke or equivalent enterprise-grade topology.
- Include:
  - PrivateLink / PSC integration
  - Subnets (public, private, restricted)
  - NSGs/NACLs
  - Firewall, egress rules, DNS design
  - Prod vs non-prod isolation
  - No public exposure of data planes

### **4. Compute, Storage & API Layer**
- Compute: container orchestration (AKS/EKS/GKE), serverless, Spark/stream jobs.
- Storage:
  - Data lake (ADLS/S3/GCS)
  - Data warehouse (Synapse/Redshift/BigQuery)
  - Operational DBs (PostgreSQL/Aurora/AlloyDB)
  - Cache (Redis)
  - Search indexing service
- API:
  - Gateway
  - WAF
  - Global load balancing layer (Front Door/CloudFront/Cloud CDN)
  - AuthN/Z, rate limiting, versioning

### **5. Load Balancing & Autoscaling**
- Global vs regional LB components.
- Pod/node autoscaling logic.
- Scaling triggers (CPU/mem, queue lag, RPS).
- Scheduled scaling windows.

### **6. CI/CD Integration**
- Source control strategy.
- Build, test, security scanning.
- Artifact management (ACR/ECR/GAR).
- GitOps or pipeline-driven deployment.
- IaC with Terraform/Bicep/Deployment Manager.
- Environment promotion (dev → test → staging → prod).

### **7. Monitoring & Logging**
- Metrics, logs, traces.
- Central logging workspace.
- Alerts for:
  - API latency/error rates
  - Event/backlog accumulation
  - Job failures
  - Data quality checks
- Security posture monitoring & SIEM.

### **8. Disaster Recovery Plan**
- Primary region vs DR/secondary region.
- RPO/RTO targets.
- Replication strategy per service.
- Automated failover steps.
- Backup strategy.
- DR testing cadence.

### **9. Service Justification**
For every cloud service selected:
- Explain WHY it was chosen.
- Include:
  - Benefits
  - Performance characteristics
  - Cost levers (autoscale, tiering, pause/resume)
  - Security or compliance advantages

### **10. Architecture Diagram (Text Form)**
Provide a **clean, structured ASCII-style description** of the end-to-end architecture that can be easily converted to a diagram.

---

## **Non-Functional Requirements**
- Availability: APIs ≥ **99.9%**
- Latency: p95 < **200 ms**
- Real-time pipeline: end‑to‑insight < **5 min**
- Security:
  - Managed identities
  - Key Vault/Secrets Manager
  - Network isolation via PrivateLink
  - E2E encryption
  - WAF + DDoS protection
- Compliance: GDPR-ready, PII tokenization, RLS/CLS
- Cost optimization:
  - Autoscale everywhere
  - Storage lifecycle policies
  - Spot/preemptible compute where possible

---

## **Constraints & Preferences**
- Follow industry patterns:
  - Hub–Spoke network
  - Lakehouse (Bronze/Silver/Gold)
  - Delta/Parquet format
- Prefer managed cloud services.
- Ensure no public endpoints for data plane.
- Include SKU/tier recommendations.
- Include AWS/Azure/GCP equivalents if only one provider is chosen.

---

## **Acceptance Criteria**
Your output MUST:
- Use all the section headings above.
- Provide explicit multi-AZ and network isolation detail.
- Include a clear DR failover sequence.
- Include CI/CD workflow with gates.
- List monitoring KPIs and alert thresholds.
- Provide justification for each service.
- Include a diagram-like text model.
- Avoid marketing language; be practical and technical.

---

## **Optional Enhancements**
- Data contracts (OpenAPI/AsyncAPI + Schema Registry)
- Feature store for ML
- Governance/catalog layer (Unity Catalog / Lake Formation / Fabric)
- Cost monitoring dashboards and budgets
