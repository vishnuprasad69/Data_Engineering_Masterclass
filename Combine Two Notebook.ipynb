```python
import nbformat

# Load the first notebook
with open("BigData_Fundamentals_Notebook.ipynb", encoding="utf-8") as f:
    nb1 = nbformat.read(f, as_version=4)

# Load the second notebook
with open("Databricks Study Notes.ipynb", encoding="utf-8") as f:
    nb2 = nbformat.read(f, as_version=4)

# Combine all cells
merged = nbformat.v4.new_notebook()
merged.cells = nb1.cells + nb2.cells

# Save the merged notebook
with open("Databricks Study Notes.ipynb", "w", encoding="utf-8") as f:
    nbformat.write(merged, f)

print("✅ Databricks Study Notes.ipynb created successfully!")
```
