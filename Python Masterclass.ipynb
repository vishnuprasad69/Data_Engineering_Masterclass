### 2. Comprehensions and Generators 💥
Comprehensions provide a concise way to create lists, dictionaries, and sets. Generators allow for memory-efficient iteration using lazy evaluation.

#### 1. List Comprehensions
List comprehensions are the Pythonic way to transform or filter data in a single line. Syntax: `[expression for item in iterable if condition]`.

```python
# Mapping: Square of numbers
squares = [x**2 for x in range(10)]

# Filtering: Only even squares
even_squares = [x**2 for x in range(10) if x % 2 == 0]

# Nested loops: Creating a coordinate grid
grid = [(x, y) for x in range(3) for y in range(2)]

print(f"Squares: {squares}")
print(f"Even Squares: {even_squares}")
print(f"Grid: {grid}")
```

#### 2. Generator Expressions
Similar to list comprehensions but use parentheses `()` instead of brackets `[]`. They use **lazy evaluation**, meaning they calculate items only when requested, saving memory.

```python
import sys

# List comprehension (stores everything in memory)
large_list = [x**2 for x in range(1000000)]

# Generator expression (calculates on the fly)
large_gen = (x**2 for x in range(1000000))

print(f"List size: {sys.getsizeof(large_list)} bytes")
print(f"Generator size: {sys.getsizeof(large_gen)} bytes")
```

#### 3. The `yield` Keyword
The `yield` keyword transforms a normal function into a generator function. This allows for creating custom iterators that maintain state between calls.

```python
def fibonacci_generator(limit):
    a, b = 0, 1
    while a < limit:
        yield a
        a, b = b, a + b

fib = fibonacci_generator(50)
print("Fibonacci sequence:", list(fib))
```

# 🐍 Python List Comprehensions: Examples and Interview Prep

## 1. Fundamentals: The Cuboid Problem
This problem demonstrates the use of a list comprehension with three nested loops and a filter condition.Goal: Given dimensions $x, y, z$ and an integer $n$, print all coordinates $[i, j, k]$ where $0 \le i \le x$, $0 \le j \le y$, $0 \le k \le z$, and $i + j + k \neq n$.

```python
# Sample Input:
x = 1
y = 1
z = 1
n = 2

# List Comprehension Solution
coordinates = [
    [i, j, k]
    for i in range(x + 1)
    for j in range(y + 1)
    for k in range(z + 1)
    if i + j + k != n
]

print(f"Dimensions (x, y, z): ({x}, {y}, {z})")
print(f"Exclusion Sum (n): {n}")
print("Resulting Coordinates:")
print(coordinates)
```

## 2. Popular List Comprehension Scenarios
These examples cover the most common use cases, illustrating mapping, filtering, nested loops, and conditional mapping.

### A. Mapping: Basic Transformation
Goal: Square every number in the list.

```python
numbers = [1, 2, 3, 4, 5]

# [ Expression for Item in Iterable ]
squared_list = [num ** 2 for num in numbers]

print(f"Original: {numbers}")
print(f"Squared: {squared_list}")
```

### B. Filtering: Selecting Elements
Goal: Keep only the even numbers.

```python
data = [11, 22, 33, 44, 55, 66]

# [ Expression for Item in Iterable if Condition ]
even_list = [item for item in data if item % 2 == 0]

print(f"Original: {data}")
print(f"Even Numbers: {even_list}")
```

### C. Nested Loops: Cross-Product
Goal: Generate all possible two-letter combinations.

```python
list_a = ['A', 'B']
list_b = ['x', 'y', 'z']

# [ Expression for OuterItem in OuterIterable for InnerItem in InnerIterable ]
combinations = [a + b for a in list_a for b in list_b]

print(f"List A: {list_a}, List B: {list_b}")
print(f"Combinations: {combinations}")
```

### D. Conditional Mapping (If/Else)

```python
data = [10, -5, 20, -1, 30]

# [ TrueValue if Condition else FalseValue for Item in Iterable ]
processed_list = [0 if num < 0 else num for num in data]

print(f"Original: {data}")
print(f"Processed: {processed_list}")
```

# 3. Object-Oriented Programming (OOP) in Python 🧱
OOP is a programming paradigm based on the concept of "objects", which can contain data (attributes) and code (methods).

## I. Core Concepts (The Three Pillars)
1. **Encapsulation**: Bundling data and methods that work on that data within a single unit (class) and restricting direct access to some components.
2. **Inheritance**: Allowing a class (child) to derive attributes and methods from another class (parent).
3. **Polymorphism**: The ability of different classes to be treated as instances of the same parent class through the same interface (e.g., different classes having the same method name).

```python
class Animal:
    def speak(self):
        pass

class Dog(Animal):  # Inheritance
    def speak(self):
        return "Woof!"

class Cat(Animal):  # Inheritance
    def speak(self):
        return "Meow!"

# Polymorphism in action
animals = [Dog(), Cat()]
for animal in animals:
    print(animal.speak())
```

## II. Classes and Instances
* `__init__`: The constructor method used to initialize an object's state.
* `self`: Represents the instance of the class.
* **Class vs Instance Attributes**: Instance attributes are unique to each object; class attributes are shared by all instances.

```python
class Employee:
    raise_amount = 1.05  # Class Attribute
    
    def __init__(self, name, pay):
        self.name = name    # Instance Attribute
        self.pay = pay
        
    def apply_raise(self):
        self.pay = int(self.pay * self.raise_amount)

emp_1 = Employee('Alice', 50000)
emp_1.apply_raise()
print(f"{emp_1.name} pay: {emp_1.pay}")
```

## III. Magic (Dunder) Methods
* `__str__`: Targeted at the end-user (readable).
* `__repr__`: Targeted at developers (unambiguous for debugging).

```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author
        
    def __repr__(self):
        return f"Book('{self.title}', '{self.author}')"
        
    def __str__(self):
        return f"{self.title} by {self.author}"

my_book = Book("1984", "George Orwell")
print(str(my_book))  # Uses __str__
print(repr(my_book)) # Uses __repr__
```

**Scenario: Bank Account System**
Create a system that handles deposits and withdrawals while protecting the balance (Encapsulation) and calculating interest (Method Inheritance).

```python
class Account:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.__balance = balance  # Private attribute (Encapsulation)
        
    def deposit(self, amount):
        self.__balance += amount
        return f"Deposited {amount}"
        
    def get_balance(self):
        return self.__balance

class SavingsAccount(Account): # Inheritance
    def add_interest(self, rate):
        interest = self.get_balance() * rate
        self.deposit(interest)
        return f"Added {interest} interest"

acc = SavingsAccount("Bob", 1000)
acc.add_interest(0.05)
print(f"Final Balance: {acc.get_balance()}")
```

### Problem 1: Private State Protection (Encapsulation)
**Scenario:** Create a `SmartDevice` class where the `battery_level` cannot be set directly above 100 or below 0. Use a property decorator to handle this validation.

```python
class SmartDevice:
    def __init__(self, battery):
        self._battery = battery
    
    @property
    def battery(self):
        return self._battery
    
    @battery.setter
    def battery(self, value):
        if 0 <= value <= 100:
            self._battery = value
        else:
            print("Invalid battery level!")

phone = SmartDevice(50)
phone.battery = 150  # Should trigger validation
print(f"Battery: {phone.battery}%")
```

### Problem 2: Multi-Level Inheritance (Method Overriding)
**Scenario:** Define a `Vehicle` class, a `Car` class, and an `ElectricCar` class. Each should have a `fuel_info()` method. `ElectricCar` should call the parent `Car` info but add its own battery capacity.

```python
class Vehicle:
    def fuel_info(self):
        return "Generic Vehicle Fuel"

class Car(Vehicle):
    def fuel_info(self):
        return "Petrol/Diesel"

class ElectricCar(Car):
    def fuel_info(self):
        base_info = super().fuel_info()
        return f"{base_info} and Electricity (75kWh)"

tesla = ElectricCar()
print(tesla.fuel_info())
```

### Problem 3: Operator Overloading (Dunder Methods)
**Scenario:** Create a `Vector` class that represents a 2D point (x, y). Overload the `+` operator so that two vectors can be added together directly.

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y
        
    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)
    
    def __repr__(self):
        return f"Vector({self.x}, {self.y})"

v1 = Vector(2, 4)
v2 = Vector(1, 3)
print(v1 + v2)
```

### 4. Abstract Base Classes (ABC)
**Scenario:** Define an abstract base class `Payment` with an abstract method `process_payment()`. Implement this in `CreditCard` and `PayPal` classes to ensure they both follow the same interface.

```python
from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def process_payment(self, amount):
        pass

class CreditCard(Payment):
    def process_payment(self, amount):
        return f"Processing credit card payment of ${amount}"

class PayPal(Payment):
    def process_payment(self, amount):
        return f"Processing PayPal payment of ${amount}"

# Usage
payments = [CreditCard(), PayPal()]
for p in payments:
    print(p.process_payment(100))
```

# 5. Iterators and Decorators ✨
This section covers two advanced Python concepts: **Iterators**, which power loops and data streams, and **Decorators**, which allow you to modify the behavior of functions dynamically.

## I. Iterators: Under the Hood of Loops

### Iterable vs. Iterator
1. **Iterable**: An object that defines the `__iter__` method (e.g., Lists, Tuples, Strings). It can produce an iterator.
2. **Iterator**: An object that defines the `__next__` method. It keeps track of its state and returns the next value one at a time.

When you use a `for` loop, Python calls `iter()` on your collection to get an iterator, and then repeatedly calls `next()` until a `StopIteration` error occurs.

```python
numbers = [1, 2, 3]
it = iter(numbers) # Get iterator

print(next(it))  # 1
print(next(it))  # 2
print(next(it))  # 3
# next(it) would now raise StopIteration
```

## II. Decorators (@ Syntax)

A **Decorator** is a function that takes another function and extends its behavior without explicitly modifying it. They are commonly used for logging, access control, and timing.

```python
def my_decorator(func):
    def wrapper():
        print("Something is happening before the function is called.")
        func()
        print("Something is happening after the function is called.")
    return wrapper

@my_decorator
def say_hello():
    print("Hello!")

say_hello()
```

### Problem 1: Custom Iterator (Countdown)
**Question:** Create a custom iterator class that counts down from a given number to zero.

```python
class Countdown:
    def __init__(self, start):
        self.current = start

    def __iter__(self):
        return self

    def __next__(self):
        if self.current < 0:
            raise StopIteration
        val = self.current
        self.current -= 1
        return val

for num in Countdown(3):
    print(num)
```

### Problem 2: Timing Decorator (Performance Testing)
**Question:** Write a decorator `@time_it` that measures and prints how long a function takes to execute.

```python
import time

def time_it(func):
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()
        print(f"{func.__name__} took {end-start:.4f} seconds")
        return result
    return wrapper

@time_it
def slow_function():
    time.sleep(1)
    print("Finished work")

slow_function()
```

# 7. Context Managers (the `with` statement) 📂
Context Managers are a Pythonic way to manage resources like files, network connections, or database locks. They ensure that resources are properly acquired and, most importantly, released—even if an error occurs.

## I. Purpose: Automatic Resource Management
Without a context manager, you must manually close files. If a crash happens before `file.close()`, the file stays open, leading to memory leaks or data corruption.

The `with` statement guarantees cleanup. It is essentially a cleaner, more robust alternative to a `try...finally` block.

```python
# The Pythonic way to handle files
with open('example.txt', 'w') as f:
    f.write("Managing resources automatically!")

# File is automatically closed here, even if the code above crashed.
```

## II. Mechanism: The Magic Methods
A class becomes a Context Manager by implementing two "dunder" methods:
1. **`__enter__(self)`**: Executed when the `with` block starts. It typically returns the resource.
2. **`__exit__(self, exc_type, exc_value, traceback)`**: Executed when the block ends. It handles closing the resource and can suppress exceptions.

```python
class MyDatabaseConnection:
    def __enter__(self):
        print("Connecting to database...")
        return "DATABASE_CONNECTION_OBJECT"

    def __exit__(self, exc_type, exc_val, exc_tb):
        print("Closing database connection...")
        # If an error occurred, exc_type will not be None
        if exc_type:
            print(f"An error was handled: {exc_val}")
        return True  # Suppresses the exception

with MyDatabaseConnection() as db:
    print(f"Using: {db}")
    # raise ValueError("Simulated crash") # Uncomment to see __exit__ handle it
```

**Question:** Write a context manager that changes the current working directory to a specific path, and automatically changes it back to the original directory once the code finishes.

```python
import os

class ChangeDir:
    def __init__(self, new_path):
        self.new_path = new_path
        self.old_path = os.getcwd()

    def __enter__(self):
        os.chdir(self.new_path)
        print(f"Entered: {os.getcwd()}")

    def __exit__(self, etype, evalue, etb):
        os.chdir(self.old_path)
        print(f"Back to: {os.getcwd()}")

with ChangeDir('/tmp'): # Use an existing path on your system
    print("Doing work in temporary folder...")
```

# 9. Multithreading vs. Multiprocessing ⚡
Python offers two main ways to run code concurrently: **Multithreading** and **Multiprocessing**. Choosing the right one depends on whether your task is waiting for external resources (I/O) or crunching numbers (CPU).

## I. Concurrency vs. Parallelism
* **Concurrency (Multithreading)**: Managing multiple tasks at the same time by switching between them quickly. It's like a waiter serving multiple tables; they aren't cooking and serving at the exact same instant, but they keep all tasks moving.
* **Parallelism (Multiprocessing)**: Executing multiple tasks at the exact same time on different CPU cores. It's like having multiple chefs in the kitchen, each cooking a different dish simultaneously.

## II. The Global Interpreter Lock (GIL)
The GIL is a mutex that allows only one thread to hold control of the Python interpreter at a time. 

### Why is this important?
1. **Impact**: Even if you have a multi-core processor, the GIL prevents multiple threads from executing Python code in parallel.
2. **The Solution**: To achieve true parallelism for **CPU-bound** tasks (like heavy math), you must use **Multiprocessing**, which creates separate memory spaces and separate Python instances for each process, bypassing the GIL.

## III. Multithreading (Best for I/O Bound Tasks)
Use threading when your program spends most of its time waiting (e.g., for a network response or disk read).

```python
import threading
import time

def io_task(name):
    print(f"Task {name} starting (waiting for 1s)...")
    time.sleep(1) # Simulates network/disk wait
    print(f"Task {name} finished.")

start = time.perf_counter()
threads = []
for i in range(5):
    t = threading.Thread(target=io_task, args=(i,))
    t.start()
    threads.append(t)

for t in threads:
    t.join()

end = time.perf_counter()
print(f"Total time with threading: {end - start:.2f} seconds")
```

## IV. Multiprocessing (Best for CPU Bound Tasks)
Use multiprocessing for tasks that require heavy computation and can be split across CPU cores.

```python
import multiprocessing
import time

def cpu_task(n):
    # Simulates a heavy calculation
    count = 0
    for i in range(n):
        count += i
    return count

if __name__ == "__main__":
    numbers = [10**7, 10**7, 10**7, 10**7]
    
    start = time.perf_counter()
    # Using a Pool of processes
    with multiprocessing.Pool() as pool:
        results = pool.map(cpu_task, numbers)
    
    end = time.perf_counter()
    print(f"Results: {results}")
    print(f"Total time with multiprocessing: {end - start:.2f} seconds")
```

# 10. Memory Management and Garbage Collection ♻️
Python manages memory automatically, so developers rarely need to allocate or deallocate memory manually. Understanding how this works helps in writing high-performance, leak-free applications.

## I. Reference Counting: The Primary Mechanism
In CPython, the main way memory is managed is through **Reference Counting**. Every object keeps track of how many references point to it.

* When an object's reference count drops to **zero**, its memory is immediately deallocated.
* You can check the reference count of an object using `sys.getrefcount()`.

```python
import sys

a = [1, 2, 3]  # Count is 1
b = a          # Count is 2

print(f"Reference count for 'a': {sys.getrefcount(a) - 1}") 

# Note: sys.getrefcount adds 1 because the function itself creates a temporary reference.
```

## II. The Generational Garbage Collector (GC)
Reference counting alone cannot handle **Circular References** (where Object A points to B, and B points back to A). Even if no one else uses them, their count never hits zero.

Python’s **Cyclic Garbage Collector** solves this by:
1. Scanning for groups of objects that only point to each other.
2. Using **Generations** (0, 1, and 2): Newer objects are scanned more frequently than older objects, which is based on the "Infant Mortality" hypothesis (most objects die young).

```python
import gc

class MyClass:
    pass

obj_a = MyClass()
obj_b = MyClass()

obj_a.other = obj_b  # A points to B
obj_b.other = obj_a  # B points to A

del obj_a
del obj_b

# At this point, the objects are 'garbage', but reference counts are still 1.
print(f"GC threshold: {gc.get_threshold()}")
print(f"Manual GC collection found: {gc.collect()} unreachable objects.")
```
