```python
 # Classical Pyramid

def classical_pyramid(n):
    for i in range(1,n+1):
        spaces = " " * (n-i)
        stars = "*" * ( 2 * i - 1 )
        print(spaces + str(stars))
classical_pyramid(10)
```

```python
def alternative_pyramid(n):
    for i in range(n):
        char = '#' if i % 2 == 0 else '*'
        print( ' ' * i + char * (2 * (n-i) - 1))
    for i in range(n-2, -1, -1):
        char = '#' if i % 2 == 0 else '*'
        print( ' ' * i + char * (2 * (n-i) - 1))


alternative_pyramid(5)
```

```python
def numbered_peak(n=5):

    for i in range(1, n+1):
        print(" "*(n-i), end="")
    
        for j in range(1, i+1):
            print(j, end="")
    
        for j in range(i-1, 0, -1):
            print(j, end="")
    
        print()
numbered_peak(5)
```

```python
def snak_matrix(n):
    num = 1
    for i in range(n):
        row = []
        for j in range(n):
            row.append(num)
            num += 1
        if i % 2 == 1:
            row.reverse()
        print(row)

snak_matrix(5)
```

```python
def check_box(row = 4, col = 4):
    for i in range (row):
        r_list = []
        for j in range(col):
            value = (i + j) % 2
            r_list.append(value)
        print(r_list)
check_box()
```

```python
def repeat_text(Text = "HELLO", length = 8, width =4):
    index = 0
    for _ in range (width):
        line = ''
        for _ in range(length):
            line += Text[index % len(Text)]
            index += 1
        print(line)

repeat_text()
```
