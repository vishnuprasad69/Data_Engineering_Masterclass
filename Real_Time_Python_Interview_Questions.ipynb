# 🚀 Senior Data Engineering & Python Interview Syllabus

## Phase 1: Logical & Visual Structural Patterns
*Focus: Mastering loops, indices, and coordinate geometry.*

### 1. Symmetrical Pyramid & Diamond Structures
* **Requirement:** Generate perfectly centered visual markers for UI alignment or terminal-based progress tracking.
* **Technical Skill:** Nested loops, string multiplication, and whitespace management.
* **Challenges:**
    1. **The Classic Pyramid:** Write a function to print a perfectly centered pyramid of `*` of height N.
    2. **The Hollow Diamond:** Generate a diamond shape of width N, but only print the border `*` characters.
    3. **Alternating Hourglass:** Print an hourglass shape of height N where rows alternate between `#` and `*`.
    4. **The Numbered Peak:** Create a pyramid where the left side counts up to the row number, and the right counts down (e.g., `12321`).
    5. **Dynamic Alignment:** Given a list of strings of varying lengths, print them vertically as a perfectly center-aligned block.
    6. **The Arrowhead:** Generate an arrowhead pattern pointing right `>` of width N, then reverse it to point left `<`.
    7. **Star of David:** Print two intersecting hollow triangles to form a symmetrical hexagram.
    8. **Concentric Diamonds:** Print a large diamond containing smaller, nested diamonds inside it.
    9. **Isometric 3D Cube:** Use slashes `/`, `\`, and underscores `_` to draw a 3D-looking cube of dynamic size N.
    10. **The Sine Wave:** Print a continuous vertical wave pattern using `*` by calculating the offset with a mathematical sine function.

### 2. Border & Hollow Shape Generation
* **Requirement:** Print frames or bounding boxes to isolate specific data areas in a CLI output.
* **Technical Skill:** Conditional logic within loops (if i == 0 or j == 0).
* **Challenges:**
    1. **The Data Frame:** Print a hollow rectangle of dimensions M x N using `-`, `|`, and `+`.
    2. **The Crossed Box:** Print a hollow square of size N x N including both diagonal lines.
    3. **The Hollow Triangle:** Print a right-angled triangle where only the hypotenuse and base are visible.
    4. **Concentric Squares:** Print N nested hollow squares alternating between `#` and `O`.
    5. **The Hexagon:** Use slashes and underscores to generate a symmetrical hexagon.
    6. **Window Pane:** Print a hollow box partitioned exactly into four equal inner quadrants by a crosshair.
    7. **The Terminal Circle:** Use the distance formula to print a hollow circle using asterisks.
    8. **Plus Sign Grid:** Generate a grid of isolated `+` signs, separated by K empty spaces.
    9. **Diamond in a Box:** Print a solid box of `#`, but carve out a hollow diamond shape perfectly centered inside.
    10. **Staircase Border:** Print a border that steps down diagonally instead of forming straight vertical lines.

### 3. Alphabetical & Numeric Sequential Patterns
* **Requirement:** Create labeled grids for mapping coordinate systems or spreadsheet-style headers.
* **Technical Skill:** ASCII conversion (`ord()`, `chr()`) and incrementing counters.
* **Challenges:**
    1. **Excel Column Mapping:** Convert an integer to its corresponding Excel column title (e.g., 1 -> A, 28 -> AB).
    2. **The Alphabet Triangle:** Print a right triangle of sequential letters.
    3. **Snake Matrix Print:** Generate an N x N grid of numbers snaking back and forth row by row.
    4. **Checkerboard Mask:** Generate an M x N 2D array where elements strictly alternate between 1 and 0.
    5. **Wrapping Text Sequencer:** Repeat a string indefinitely, breaking it exactly at K characters per line.
    6. **Boustrophedon Print:** Print an N x N numeric grid where row 1 goes left-to-right, row 2 goes right-to-left, etc.
    7. **Look-and-Say Sequence:** Generate the N-th line of the Conway sequence.
    8. **Alphanumeric Spiral:** Print a grid that spirals inward, strictly alternating between letters and numbers.
    9. **Roman Numeral Grid:** Given N, print a grid counting up to N^2 using only valid Roman numerals.
    10. **Palindromic Number Diamond:** Print a diamond where the center vertical axis acts as the peak.

### 4. Pascal’s Triangle & Mathematical Progressions
* **Requirement:** Generate a hierarchical structure where each row depends on the previous row's values.
* **Technical Skill:** Combinatorics logic and list-based state retention.
* **Challenges:**
    1. **Pascal Generator:** Return the first N rows of Pascal’s Triangle as a list of lists.
    2. **Targeted Pascal:** Find the K-th element in the N-th row of Pascal's Triangle in O(N) time.
    3. **Fibonacci Pyramid:** Print a centered pyramid sequencing through the Fibonacci numbers.
    4. **Floyd’s Triangle:** Generate a right-angled triangle filled with consecutive integers.
    5. **Geometric Growth Grid:** Create a grid where each cell is the sum of the cell directly above and left.
    6. **Catalan Numbers:** Generate the first N Catalan numbers, formatting them visually into a mountain structure.
    7. **Eulerian Numbers:** Calculate and print the Eulerian numbers triangle.
    8. **Minimum Path Sum Triangle:** Find the minimum path sum from top to bottom in a triangle array.
    9. **Prime Number Pyramid:** Generate a pyramid strictly using the sequence of prime numbers.
    10. **Collatz Formatter:** Compute a Collatz conjecture sequence and print it as a visually escalating bar chart.

## Phase 2: Matrix & 2D Grid Manipulations
*Focus: Navigating and transforming multi-dimensional data.*

### 5. Spiral Matrix Traversal
* **Requirement:** Read or write data in a continuous inward loop.
* **Technical Skill:** Boundary management and 4-directional state switching.
* **Challenges:**
    1. **Spiral Extraction:** Return all elements of an M x N matrix in a 1D array traversing clockwise.
    2. **Spiral Generation:** Generate an N x N matrix filled with elements from 1 to N^2 in a clockwise spiral.
    3. **Counter-Clockwise Spiral:** Traverse a given matrix in a counter-clockwise spiral pattern.
    4. **Spiral Center:** Find the exact center element of a spiral-filled matrix without generating the matrix itself.
    5. **Spiral String Decoding:** Read characters in a spiral order from a grid to decode a hidden string message.
    6. **Diagonal Spiral:** Traverse a matrix in a spiral, strictly moving in diagonal lines.
    7. **Double Pointer Spiral:** Generate an N x N spiral using two simultaneous pointers.
    8. **String to Grid Unfolding:** Take a 1D string and wrap it into an M x N grid following a clockwise spiral.
    9. **Matrix Peeling:** Return the outermost layer of a matrix, remove it, and repeat until empty.
    10. **Expanding Ripple:** Traverse the matrix outward in expanding concentric squares from a center coordinate.

### 6. Matrix Rotation & Transposition
* **Requirement:** Rotate a dataset by 90 or 180 degrees.
* **Technical Skill:** Index swapping and list comprehensions.
* **Challenges:**
    1. **In-Place Rotation:** Rotate an N x N matrix 90 degrees clockwise without allocating a new 2D array.
    2. **The Transpose:** Transpose an M x N matrix (rows become columns).
    3. **180-Degree Image Flip:** Flip a matrix representing image pixels horizontally, then vertically.
    4. **Shift and Wrap:** Shift all elements right by K steps, wrapping elements to the next row.
    5. **Ring Rotation:** Rotate the outer border of an N x N matrix clockwise by 1 step.
    6. **Sub-Grid Rotation:** Rotate a specific K x K sub-grid at coordinate (x,y) by 90 degrees.
    7. **Shear Transformation:** Apply a horizontal shear where each row is shifted right by its row index.
    8. **Secondary Diagonal Reflection:** Transpose a matrix across its anti-diagonal.
    9. **Concentric Ring Rotation:** Rotate the outermost ring clockwise by 1, the next counter-clockwise by 1.
    10. **In-Place Shift:** Shift all elements in the matrix right by K spaces entirely in-place.

### 7. Diagonal & Anti-Diagonal Extraction
* **Requirement:** Isolate specific cross-sections of a grid for correlation analysis.
* **Technical Skill:** Linear equations of indices.
* **Challenges:**
    1. **Primary and Secondary Diagonals:** Extract and print only the elements on the two main diagonals.
    2. **Zig-Zag Traversal:** Return elements traversed in a zig-zag (diagonal) order.
    3. **Matrix Diagonal Sort:** Sort every diagonal running top-left to bottom-right in ascending order.
    4. **Toeplitz Matrix Validation:** Check if every diagonal from top-left to bottom-right contains the same elements.
    5. **Diagonal Sum Difference:** Calculate the absolute difference between the sum of the primary and secondary diagonals.
    6. **N-Way Tic-Tac-Toe:** Check if any row, column, primary, or anti-diagonal is completely filled with the same character.
    7. **Longest Consecutive Diagonal:** Find the longest sequence of identical numbers on any diagonal.
    8. **Bounded Diagonal Sum:** Sum only the primary diagonal of a specific bounded sub-grid.
    9. **Independent Diagonal Sort:** Sort every single diagonal parallel to the main diagonal without mixing numbers.
    10. **Mirror Across Main:** Swap element (i, j) with (j, i) only if (i, j) is greater than (j, i).

### 8. Sparse Matrix Compression
* **Requirement:** Create a structural representation that only stores "active" coordinates.
* **Technical Skill:** Dictionary-based coordinate mapping.
* **Challenges:**
    1. **Dense to COO:** Convert a dense 2D list into a Coordinate List (COO) sparse matrix representation.
    2. **Sparse Addition:** Add two sparse matrices represented as dictionaries.
    3. **Sparse Vector Multiplication:** Multiply a sparse matrix (DOK format) by a 1D dense vector.
    4. **Reconstruct Dense Matrix:** Given dimensions and a sparse dictionary, reconstruct the full dense matrix.
    5. **Sparse Transposition:** Transpose a sparse dictionary matrix without converting it back to a dense matrix.
    6. **Dense to CSR Conversion:** Convert a 2D list into Compressed Sparse Row (CSR) format.
    7. **CSR Matrix Multiplication:** Multiply two CSR matrices without converting them to dense arrays.
    8. **Sparse Flood Fill:** Implement a flood fill algorithm on a grid represented entirely as a sparse coordinate dictionary.
    9. **3D Sparse Tensor:** Compress a 3D grid full of zeros into a sparse dictionary and return a 2D "slice".
    10. **Sparse Submatrix Extraction:** Extract a bounded K x K submatrix directly from a sparse dictionary.

## Phase 3: String & Textual Structural Logic
*Focus: Reshaping and validating unstructured text.*

### 9. Palindrome & Anagram Structural Validation
* **Requirement:** Determine if a text structure is a mirror image or a character permutation.
* **Technical Skill:** Slicing, frequency counters, and two-pointer techniques.
* **Challenges:**
    1. **Anagram Grouper:** Group a list of strings into sub-lists of anagrams.
    2. **Palindrome Permutation:** Check if a string's characters can be rearranged to form a palindrome.
    3. **Longest Palindromic Substring:** Find the longest contiguous substring that reads the same forwards and backwards.
    4. **Anagram Indices:** Find all starting indices of anagrams of string P within string S.
    5. **Minimum Insertions:** Find the minimum characters to insert to make a string a valid palindrome.
    6. **Min-Append Palindrome:** Find the minimum characters to append to the end to make it a palindrome.
    7. **Group Shifted Strings:** Group strings that are anagram-like shifts of each other.
    8. **Wildcard Anagram:** Check if two strings are anagrams where `?` acts as a wildcard.
    9. **Palindromic Partitioning:** Return all possible ways to slice a string such that every slice is a valid palindrome.
    10. **Shortest Anagram Window:** Find the shortest contiguous window in string S containing all characters of T.

### 10. String Expansion & Compression (Run-Length)
* **Requirement:** Take a string like `a3b2` and expand it to `aaabb`, or vice versa.
* **Technical Skill:** Character iteration and stateful counting.
* **Challenges:**
    1. **Basic RLE:** Compress a string using Run-Length Encoding.
    2. **Nested Expansion:** Decode a string formatted as `3[a2[c]]` into `accaccacc`.
    3. **Iterative Expansion:** Expand `a3b2c1` to `aaabbbc` without external libraries.
    4. **Streaming Compression:** Write a generator yielding run-length encoded tuples from a string stream.
    5. **Length Prediction:** Calculate the total length of an expanded string from RLE format without allocating memory.
    6. **LZ77 Sliding Window:** Implement basic LZ77 compression logic using a sliding window.
    7. **2D Run-Length Encoding:** Compress a 2D pixel array by applying RLE across row boundaries.
    8. **Regex Expansion:** Expand `ab(c2)3` into `abccabccabcc` (handling grouped parentheses with multipliers).
    9. **Compressed Intersection:** Find the intersection length of two massive RLE strings without expanding them.
    10. **RLE Block Swapping:** Swap the i-th compressed block with the j-th block in an RLE string.

### 11. Word Wrapping & Text Justification
* **Requirement:** Format a long string into a fixed-width block.
* **Technical Skill:** Length calculations and greedy word distribution.
* **Challenges:**
    1. **Basic Word Wrap:** Break a string into lines of maximum length K without splitting words.
    2. **Full Justification:** Pack words into lines, inserting spaces so margins align perfectly.
    3. **No Adjacent Duplicates:** Reformat a string so no two adjacent characters are the same.
    4. **The Typewriter:** Insert a newline exactly two spaces after every period, exclamation point, or question mark.
    5. **Longest Word Centering:** Center all words in a list relative to the width of the longest word.
    6. **Columnar Terminal Layout:** Format words into K columns of equal width.
    7. **ASCII Image Wrapping:** Wrap text smoothly around an ASCII "image" box.
    8. **Hyphenation Wrap:** Wrap text using a dictionary to split overly long words with a hyphen at valid syllables.
    9. **Vertical & Horizontal Center:** Place a string perfectly in the center of an M x N character grid.
    10. **CSV to Markdown Table:** Convert a raw CSV string into a perfectly aligned Markdown table.

### 12. Nested Bracket & Expression Balancing
* **Requirement:** Ensure code or JSON structures are closed correctly.
* **Technical Skill:** Stack-based (LIFO) tracking.
* **Challenges:**
    1. **Valid Parentheses:** Determine if a string of `()`, `{}`, and `[]` is validly closed and nested.
    2. **Max Nesting Depth:** Find the maximum depth of nested parentheses in an expression.
    3. **Minimum Removals:** Remove the minimum number of invalid parentheses to make a string valid.
    4. **String Calculator:** Evaluate a basic math expression string containing numbers, operators, and brackets.
    5. **Reverse Substrings in Brackets:** Reverse the substrings enclosed in matching brackets from the inside out.
    6. **HTML Tag Validator:** Parse HTML to ensure every `<tag>` has a matching `</tag>`, returning the first unbalanced one.
    7. **Reverse Polish Notation:** Evaluate an arithmetic expression provided in RPN using a stack.
    8. **Manual JSON Decoder:** Parse a nested JSON-like string into native Python objects without the `json` library.
    9. **Deepest Bracket Content:** Return the text located inside the deepest level of nested parentheses.
    10. **Generate Valid Brackets:** Generate all possible combinations of N pairs of perfectly balanced parentheses.

## Phase 4: Data Engineering & Object Patterns
*Focus: Solving real-world data flow and structural challenges.*

### 13. Flattening Complex Hierarchies
* **Requirement:** Convert deeply nested structures into a single flat structure.
* **Technical Skill:** Recursion and yield generators.
* **Challenges:**
    1. **Recursive List Flattening:** Completely flatten a list containing arbitrarily deep nested lists.
    2. **JSON Dot-Notation:** Flatten a nested JSON dictionary into a single-level dictionary representing paths.
    3. **Un-flatten JSON:** Reconstruct a deeply nested JSON object from a dot-notation dictionary.
    4. **Deep Key Extraction:** Yield all values associated with a specific target key in a nested hierarchy.
    5. **Matrix Flattening:** Manually flatten an N-dimensional matrix into a 1D array.
    6. **File System Paths:** Flatten a nested dictionary representing a file system into a list of absolute paths.
    7. **XML to Tabular:** Flatten an XML-like hierarchical element tree into a strictly 2D tabular format.
    8. **The Un-Flattener:** Reconstruct a nested dictionary from a list of absolute file paths.
    9. **Tree to Heap Array:** Flatten a Binary Tree object into a 1D array using Heap indexing formulas.
    10. **Depth-Limited Flattening:** Flatten a nested list but stop when reaching depth K.

### 14. Data Deduplication & Unique Structuring
* **Requirement:** Filter a stream of records to ensure only unique entries remain.
* **Technical Skill:** Hashing and Set-based membership testing.
* **Challenges:**
    1. **Ordered Dict Deduplication:** Remove duplicates from a list of dicts based on a key while preserving order.
    2. **First Non-Repeating Character:** Find the first non-repeating character in a continuous string in O(N) time.
    3. **Massive File Intersection:** Find intersecting lines between two 50GB unsorted log files.
    4. **Latest Event Filter:** Filter session events to return only the absolute latest event for each unique user.
    5. **O(1) Set with Random:** Design a data structure supporting insert, remove, and getRandomElement in O(1) time.
    6. **Bloom Filter Concept:** Implement a probabilistic Bloom Filter array to filter duplicates from 10 million IPs.
    7. **Duplicate Subtrees:** Search a JSON object and return all nested dictionaries that appear identically more than once.
    8. **Fuzzy Deduplication:** Combine strings with a Levenshtein distance of 1 (a typo) as duplicates.
    9. **Top K Unique Stream:** Maintain a structure returning the top K most frequent unique items from an infinite stream.
    10. **Time-Window Deduplication:** Discard an event only if the exact same event occurred within the trailing 10-minute window.

### 15. Chronological Data Sorting & Merging
* **Requirement:** Combine pre-sorted lists of timestamps into one.
* **Technical Skill:** Merge-sort logic and pointer comparison.
* **Challenges:**
    1. **Merge K Sorted Lists:** Merge K separate, pre-sorted lists of timestamps into a single master list.
    2. **Merge Overlapping Intervals:** Merge all overlapping intervals into contiguous blocks.
    3. **Meeting Rooms:** Determine the minimum conference rooms required so no meeting intervals conflict.
    4. **External Sort Merge:** Merge two sorted files on disk into a third file without loading either entirely into memory.
    5. **Chronological Validation:** Verify if 10 million timestamps strictly follow chronological order.
    6. **K-Way Merge with Heap:** Efficiently merge K massive, pre-sorted log files using Python's `heapq`.
    7. **Calendar Availability:** Calculate overlapping free intervals where all employees in a list can meet.
    8. **Server Uptime:** Calculate absolute total uptime from a log of overlapping start/stop timestamps.
    9. **Fuzzy Timestamp Sort:** Chronologically sort a list of logs with varying timestamp formats.
    10. **Circular Buffer Insertion:** Implement a fixed-size circular buffer to efficiently insert chronologically ordered events.

### 16. Sliding Window Aggregations
* **Requirement:** Calculate moving averages or stats in a continuous stream.
* **Technical Skill:** Deque (double-ended queue) and fixed-length windowing.
* **Challenges:**
    1. **Max Subarray Sum:** Find the maximum sum of any contiguous subarray of size K.
    2. **Streaming Moving Average:** Create a class to ingest an integer stream and return the moving average.
    3. **Longest Substring with K Distinct:** Find the length of the longest substring containing at most K distinct characters.
    4. **Sliding Window Median:** Calculate the exact median value for every sliding window of size K.
    5. **Fraud Detection Window:** Detect if an event occurs 3 or more times within a rolling 5-minute window.
    6. **Deque Max Window:** Use `collections.deque` to find the maximum value in every sliding window of size K.
    7. **Exponential Moving Average:** Write a generator yielding the EMA of a streaming sequence.
    8. **Spike Detection:** Trigger an alert if the current value exceeds 200% of the sliding window average of the previous 100 values.
    9. **Rolling Distinct Count:** Calculate the exact number of unique IP addresses within a sliding 1-hour window.
    10. **Sliding Window Mode:** Continuously update the mode as a window of size K slides across an array.

### 17. Group-By / Frequency Mapping
* **Requirement:** Categorize a raw list of items into a summary structure.
* **Technical Skill:** `collections.Counter` and dictionary grouping.
* **Challenges:**
    1. **Category Aggregation:** Group transaction dicts by "Category" and return the total sum of "Amount".
    2. **Top K Frequent Elements:** Find top K most frequent words in a massive array using a Priority Queue.
    3. **Anagram Length Grouper:** Group an array of strings by string length using `collections.defaultdict`.
    4. **Time-Series Grouping:** Count distinct IP addresses per hour from a log of timestamps and IPs.
    5. **Pure Python Map-Reduce:** Implement map-reduce logic using `map()` and `functools.reduce()`.
    6. **Manual Multi-Aggregator:** Write a custom `group_by` function handling sum for one key and mean for another.
    7. **Time-Bucket Hashing:** Group exact Unix timestamps into 15-minute intervals.
    8. **Hierarchical Grouping:** Group user dicts by "Country", then sub-group by "State".
    9. **Boyer-Moore Majority:** Find an element appearing more than N/3 times in an array using O(1) space.
    10. **Inverse Dictionary Map:** Invert a dictionary where values are lists so keys are list items mapping back to the original key.

### 18. Linked Data Navigation (Pointers)
* **Requirement:** Connect separate data nodes where each item knows the next.
* **Technical Skill:** Class-based node linking (Linked List logic).
* **Challenges:**
    1. **Cycle Detection:** Detect if a cycle exists in a linked list using Floyd’s Tortoise and Hare.
    2. **Middle of the Linked List:** Find the exact middle node in a single pass using slow/fast pointers.
    3. **In-Place Reversal:** Reverse a singly linked list in-place without allocating new nodes.
    4. **Merge Sorted Linked Lists:** Merge two sorted linked lists by splicing nodes.
    5. **Flatten Multi-Level List:** Flatten a doubly linked list where nodes have `next`, `prev`, and `child` pointers.
    6. **List Intersection:** Find the exact node where two independent singly linked lists merge.
    7. **LRU Eviction Pointer:** Implement Doubly Linked List + Hash Map pointer logic for an LRU cache.
    8. **Reorder List:** Reorder a linked list from L1 -> L2 -> L3 -> L4 to L1 -> L4 -> L2 -> L3 in place.
    9. **Deep Copy with Random Pointer:** Deep copy a linked list where nodes have standard `next` and arbitrary `random` pointers.
    10. **Remove N-th from End:** Remove the N-th node from the end of a list in a single pass.

### 19. File System/Tree Traversal
* **Requirement:** Search through a nested folder structure to find specific file types.
* **Technical Skill:** Depth-First Search (DFS) or Breadth-First Search (BFS).
* **Challenges:**
    1. **Print Directory Tree:** Recursively traverse a file system dictionary and print a visually indented tree.
    2. **Find File by Extension:** Traverse a nested dictionary to return paths of all files ending in `.log`.
    3. **Max Depth of Binary Tree:** Calculate the maximum depth of a hierarchical tree structure.
    4. **Level Order Traversal:** Print the nodes of a tree level-by-level (BFS).
    5. **Invert Binary Tree:** Swap the left and right children of all nodes in a binary tree.
    6. **Recursive Directory Size:** Calculate the total size of a specified parent directory from a nested dictionary.
    7. **Lowest Common Ancestor:** Find the lowest common manager of two specific employees in an org chart tree.
    8. **N-Ary Tree Serialization:** Serialize and deserialize an N-ary tree to and from a single string.
    9. **Target Path Sum:** Find all root-to-leaf paths in a tree where node sums equal a target value.
    10. **Trie Builder:** Implement a Prefix Tree to ingest a word dictionary and support autocomplete queries.

### 20. Memory-Efficient Streaming (Generators)
* **Requirement:** Process massive files without loading them into RAM.
* **Technical Skill:** Lazy evaluation and the `yield` keyword.
* **Challenges:**
    1. **Line-by-Line Log Parser:** Write a generator to parse a 10GB log file, yielding only lines containing "ERROR".
    2. **Fibonacci Generator:** Create a generator yielding the Fibonacci sequence infinitely.
    3. **Moving Window Generator:** Yield a sliding window list of K items from an infinite stream.
    4. **Custom Range Generator:** Recreate Python's built-in `range()` function as a custom generator using `yield`.
    5. **Batch Yielding:** Read an API stream and yield the data in batches of 100 records at a time.
    6. **Chunked Batch Processor:** Consume an infinite stream and yield exact 1000-item chunks for database insertion.
    7. **Generator Pipeline:** Chain three generators (read, filter, transform) to demonstrate lazy data flow.
    8. **Infinite Prime Generator:** Yield prime numbers infinitely with a mechanism to break the loop.
    9. **Python `tail -f`:** Mimic Unix `tail -f` by yielding new lines appended to a live log file.
    10. **Massive CSV Merge:** Merge two 50GB sorted CSV files line-by-line into a new file using generators.
