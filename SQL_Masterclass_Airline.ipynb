# Chapter 1: Joins & Set Operations 🧩
## 1. Detailed Module Content 📘
### Core Concepts
* **INNER JOIN**: Returns records that have matching values in both tables. Essential for linking confirmed passengers to flights.
* **LEFT (OUTER) JOIN**: Returns all records from the left table, and the matched records from the right table. Critical for finding flights with *no* passengers.
* **RIGHT (OUTER) JOIN**: Returns all records from the right table, and the matched records from the left table. Less common, often swapped for Left Join.
* **FULL (OUTER) JOIN**: Returns all records when there is a match in either left or right table. Useful for comparing two lists of airports to see discrepancies.
* **CROSS JOIN**: Returns the Cartesian product (all combinations). Used for generating all possible route combinations.
* **SELF JOIN**: A regular join, but the table is joined with itself. Vital for hierarchical data like Pilot/Co-Pilot assignments from an 'Employees' table.

### Airline Realistic Example
Imagine linking a `Boarding_Pass` to a `Passenger`. If you use an **Inner Join**, you only see checked-in passengers. If you use a **Left Join** (from Passenger), you see everyone who bought a ticket, even if they haven't checked in (Boarding Pass is NULL).


## 2. Top 10 Major Area (Theory) Questions 🧠
**Q1:** What is the difference between WHERE and ON in a JOIN clause?

**A:** `ON` defines the relationship between the tables during the join process. `WHERE` filters the result set *after* the join has logically occurred (though optimizers may rearrange this).

---**Q2:** When would you use a CROSS JOIN in an airline context?

**A:** To generate a matrix of every `Aircraft` type against every `Airport` to analyze theoretical landing compatibility.

---**Q3:** Explain a scenario where a LEFT JOIN is preferable to an INNER JOIN.

**A:** Reporting on all scheduled flights, including those that have zero bookings yet. An Inner Join would hide empty flights.

---**Q4:** What happens if you join two tables on a column containing NULLs?

**A:** NULLs do not match each other in SQL joins. Rows with NULL keys will not appear in an INNER JOIN result.

---**Q5:** How does a Self-Join help in crew management?

**A:** It allows matching an employee (Pilot) with another employee (Co-Pilot) from the same `Crew` table based on a shared `Flight_ID`.

---**Q6:** What is a Cartesian Product and how do you avoid it?

**A:** It occurs when you join tables without a valid condition, multiplying row counts (A x B). Avoid it by always ensuring proper primary/foreign key constraints in the `ON` clause.

---**Q7:** Can you join more than two tables? Is there a limit?

**A:** Yes, you can chain joins. There is a technical limit depending on the DB (e.g., PostgreSQL is unlimited in practice, SQL Server is 256), but performance degrades with complexity.

---**Q8:** What is the performance impact of joining on non-indexed columns?

**A:** The database must perform full table scans to find matches, which is extremely slow. Joining on Indexed columns (Foreign Keys) is standard practice.

---**Q9:** Difference between UNION and JOIN?

**A:** JOIN combines columns horizontally from different tables. UNION combines rows vertically from similar result sets.

---**Q10:** What is a Natural Join?

**A:** A join that automatically links tables based on columns with the same name. It is generally discouraged in production due to the risk of unintended matches.

# Chapter 2: Aggregations & Grouping 📊
## 1. Detailed Module Content 📘
### Core Concepts
* **GROUP BY**: Groups rows that have the same values into summary rows. Essential for reporting (e.g., 'Flights per Day').
* **HAVING**: Filters the *groups* created by GROUP BY. (e.g., 'Days with > 100 flights'). Distinct from WHERE, which filters rows.
* **COUNT()**: Counts rows. `COUNT(*)` counts all, `COUNT(col)` ignores NULLs.
* **SUM() / AVG()**: Mathematical aggregates.
* **MIN() / MAX()**: Extremes.

### Airline Realistic Example
Calculating the Load Factor. You need to count the number of passengers per flight and divide by the aircraft capacity. This requires grouping by `Flight_ID`.

## 2. Top 10 Major Area (Theory) Questions 🧠
**Q1:** What is the order of execution for a query with WHERE, GROUP BY, and HAVING?

**A:** FROM -> WHERE -> GROUP BY -> HAVING -> SELECT -> ORDER BY.

---**Q2:** Can you use an alias defined in the SELECT clause inside the HAVING clause?

**A:** Generally NO (in standard SQL), because SELECT runs after HAVING. Some DBs (MySQL) allow it, but it's bad practice.

---**Q3:** Difference between COUNT(*) and COUNT(column_name)?

**A:** `COUNT(*)` counts total rows including NULLs. `COUNT(col)` counts only non-NULL values in that column.

---**Q4:** Why can't you use non-aggregated columns in the SELECT list if they aren't in GROUP BY?

**A:** Because the database wouldn't know which specific row's value to display for that column in the collapsed group.

---**Q5:** How do you find the top 3 highest revenue flights?

**A:** Aggregate SUM(revenue), GROUP BY flight, ORDER BY revenue DESC, LIMIT 3.

---**Q6:** What does GROUP BY 1, 2 mean?

**A:** It groups by the 1st and 2nd columns specified in the SELECT clause. Convenient but risky if column order changes.

---**Q7:** Can you Group By a calculated field?

**A:** Yes, e.g., `GROUP BY EXTRACT(YEAR FROM flight_date)` to aggregate by year.

---**Q8:** How to filter for groups that have at least 5 records?

**A:** Use `HAVING COUNT(*) >= 5`.

---**Q9:** What is GROUP_CONCAT (or STRING_AGG)?

**A:** An aggregate function that concatenates string values from a group into a single string (e.g., listing all passenger names in one cell).

---**Q10:** When is DISTINCT necessary in an aggregate function?

**A:** When you want to count unique items, e.g., `COUNT(DISTINCT airport_code)` to see how many unique airports were visited.

# Chapter 3: Window Functions 🪟
## 1. Detailed Module Content 📘
### Core Concepts
* **OVER()**: Defines the window (set of rows) to perform the calculation on.
* **PARTITION BY**: Divides the window into partitions (e.g., partition by `Airline` to rank planes within each airline separately).
* **ORDER BY**: Defines the logical order within the window (e.g., by `Flight_Date` for running totals).
* **Ranking**: `ROW_NUMBER()` (Unique), `RANK()` (Skips on tie), `DENSE_RANK()` (No skip).
* **Offset**: `LEAD()` (Next row), `LAG()` (Previous row).

### Airline Realistic Example
Identifying a flight delay trend. Using `LAG()`, you can compare the delay of the current flight with the delay of the *previous* flight by the same aircraft to see if delays are compounding.

## 2. Top 10 Major Area (Theory) Questions 🧠
**Q1:** Main difference between Window Functions and Group By?

**A:** Group By collapses rows into a single summary. Window Functions keep the original rows but add a calculated column based on the set.

---**Q2:** What is the default window frame if ORDER BY is specified?

**A:** `RANGE BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW` (Running Total).

---**Q3:** When would you use DENSE_RANK() over RANK()?

**A:** When you want consecutive ranking numbers even after a tie (e.g., 1, 2, 2, 3 instead of 1, 2, 2, 4). Useful for 'Top 10' lists where you don't want gaps.

---**Q4:** Explain PARTITION BY in the context of specific Airlines.

**A:** It resets the calculation counter every time the Airline ID changes. Rank #1 Delta pilot, Rank #1 United pilot, etc.

---**Q5:** What does NTILE(4) do?

**A:** Divides the result set into 4 equal buckets (Quartiles). Useful for segmenting passengers into pricing tiers.

---**Q6:** Can Window Functions be used in the WHERE clause?

**A:** No, because they are calculated after the WHERE clause filters rows. You must use a CTE or Subquery.

---**Q7:** Purpose of LAG()?

**A:** Accesses data from a previous row in the same result set without a self-join.

---**Q8:** How to calculate a Running Total?

**A:** `SUM(col) OVER (ORDER BY date)`.

---**Q9:** What is the 'Frame Clause' in a window function?

**A:** It defines the specific subset of rows within the partition (e.g., `ROWS BETWEEN 1 PRECEDING AND 1 FOLLOWING`).

---**Q10:** Difference between ROW_NUMBER and RANK when values are identical?

**A:** ROW_NUMBER assigns unique arbitrary integers (1, 2). RANK assigns the same integer (1, 1).

# Chapter 4: Subqueries & CTEs 🏗️
## 1. Detailed Module Content 📘
### Core Concepts
* **CTE (Common Table Expression)**: Defined with `WITH`. Acts as a temporary view for the query duration. Improves readability.
* **Subquery (Scalar)**: Returns a single value. Can be used in SELECT list.
* **Subquery (Table)**: Returns a table. Used in FROM clause.
* **Correlated Subquery**: References columns from the outer query. Executes once for *every* row. Performance killer.

### Airline Realistic Example
Calculating 'High Value Flights'. First, define a CTE that calculates the average revenue of *all* flights. Then, select specific flights from the main table where their revenue > that CTE value.

## 2. Top 10 Major Area (Theory) Questions 🧠
**Q1:** Why prefer CTEs over Subqueries?

**A:** Readability (linear logic vs nested) and Reusability (can reference CTE multiple times in one query).

---**Q2:** What is a Recursive CTE?

**A:** A CTE that references itself. Essential for hierarchical data like Organization Charts (Manager -> Employee) or Multi-leg flight paths.

---**Q3:** When does a Subquery run in the execution order?

**A:** Non-correlated subqueries usually run once *before* the outer query. Correlated subqueries run *during* the outer query loop.

---**Q4:** Can you use a CTE in a DELETE statement?

**A:** Yes, useful for identifying rows to delete (e.g., deleting duplicate passengers).

---**Q5:** What is the scope of a CTE?

**A:** It exists only for the execution of the single SELECT/INSERT/UPDATE/DELETE statement immediately following it.

---**Q6:** Difference between `IN` and `EXISTS` with subqueries?

**A:** `IN` checks values in a list (better for small lists). `EXISTS` returns true/false if subquery returns rows (better for large subqueries, stops at first match).

---**Q7:** Can a CTE reference another CTE?

**A:** Yes, you can chain them: `WITH CTE1 AS (...), CTE2 AS (SELECT * FROM CTE1...)`.

---**Q8:** What is a Scalar Subquery?

**A:** A subquery that returns exactly one row and one column (a single value).

---**Q9:** How to optimize a Correlated Subquery?

**A:** Rewrite it as a JOIN or use Window Functions if possible.

---**Q10:** Limit of CTEs?

**A:** No hard limit, but too many can make the query plan complex.

# Chapter 5: Data Types & Conversion 🔢
## 1. Detailed Module Content 📘
### Core Concepts
* **Date/Time**: `TIMESTAMP` (Date+Time), `DATE`, `TIME`, `INTERVAL`. Timezones are crucial in aviation.
* **String**: `CHAR` (Fixed), `VARCHAR` (Variable), `TEXT` (Unlimited).
* **Numeric**: `INT`, `BIGINT` (for IDs), `DECIMAL`/`NUMERIC` (for Currency - never use Float for money!).
* **JSON**: Storing unstructured data like 'Meal Preferences' or 'Seat Configuration'.

### Airline Realistic Example
A flight departs London at 14:00 GMT and arrives in New York at 17:00 EST. You must use `TIMESTAMPTZ` (Timestamp with Timezone) to calculate the actual duration (8 hours), otherwise 17:00 - 14:00 looks like 3 hours.

## 2. Top 10 Major Area (Theory) Questions 🧠
**Q1:** Why use DECIMAL over FLOAT for ticket prices?

**A:** FLOAT is approximate (floating point math errors). DECIMAL is exact, required for financial accuracy.

---**Q2:** Difference between CHAR(3) and VARCHAR(3) for Airport Codes?

**A:** CHAR(3) always reserves 3 bytes (padded). VARCHAR(3) uses only necessary space. For Airport Codes (always 3 chars), CHAR is slightly faster/efficient.

---**Q3:** How does PostgreSQL handle 'Timestamp with Timezone'?

**A:** It converts the input to UTC for storage and converts back to the client's timezone for display.

---**Q4:** What is an INTERVAL type?

**A:** Represents a duration of time (e.g., '2 hours 30 mins') rather than a point in time.

---**Q5:** How to extract the Year from a Date?

**A:** `EXTRACT(YEAR FROM column)` or `YEAR(column)`.

---**Q6:** Handling JSON in SQL?

**A:** Modern SQL (Postgres/MySQL) supports JSON types. query: `data->>'key'`.

---**Q7:** What is the result of '2023-01-01' + INTERVAL '1 day'?

**A:** '2023-01-02'.

---**Q8:** Best type for a Ticket Number (e.g., '001-234')?

**A:** VARCHAR. Don't use INT because of leading zeros and hyphens.

---**Q9:** Implicit vs Explicit Casting?

**A:** Implicit: DB does it auto (INT + FLOAT = FLOAT). Explicit: User forces it (`CAST(col AS DATE)`).

---**Q10:** What is UUID?

**A:** Universally Unique Identifier. Better than serial INT for distributed systems (prevents ID collision).

# Chapter 6: DDL & DML Operations 🛠️
## 1. Detailed Module Content 📘
### Core Concepts
* **DDL (Data Definition Language)**: Commands that define the structure. `CREATE`, `ALTER`, `DROP`, `TRUNCATE`.
* **DML (Data Manipulation Language)**: Commands that manipulate data. `INSERT`, `UPDATE`, `DELETE`.
* **Constraints**: Rules enforced on data columns. `PRIMARY KEY` (Unique ID), `FOREIGN KEY` (Link to another table), `UNIQUE`, `NOT NULL`, `CHECK` (Validate value range).
* **UPSERT**: A combination of UPDATE and INSERT. If the row exists, update it; otherwise, insert a new one. Crucial for loading data pipelines without duplicates.

### Airline Realistic Example
When a flight lands, the 'Actual Arrival Time' needs to be recorded. If the row for that flight exists in the `Flight_Stats` table, update it. If it's a new record (perhaps a diverted flight created on the fly), insert it. This is a classic **UPSERT** scenario.

## 2. Top 10 Major Area (Theory) Questions 🧠
**Q1:** Difference between DELETE and TRUNCATE?

**A:** DELETE removes rows one by one (logs each deletion, can have WHERE clause, slower). TRUNCATE deallocates the entire table's pages (minimal logging, no WHERE clause, very fast, resets Identity/Serial counters).

---**Q2:** What is a Foreign Key constraint?

**A:** A rule that ensures values in one column match values in the Primary Key of another table, maintaining Referential Integrity (e.g., A booking cannot exist for a non-existent Flight ID).

---**Q3:** What happens if you DROP a table that other tables depend on?

**A:** The DB will raise an error unless you use `CASCADE`, which deletes the dependent objects too (very dangerous).

---**Q4:** Difference between ALTER TABLE and UPDATE?

**A:** ALTER modifies the structure (columns, types). UPDATE modifies the actual data rows.

---**Q5:** What is a CHECK constraint?

**A:** It validates that data entered meets a specific boolean condition (e.g., `CHECK (price > 0)`).

---**Q6:** Explain `ON DELETE CASCADE`.

**A:** If a parent record is deleted, all child records referencing it are automatically deleted (e.g., Delete a Flight -> Delete all its Bookings).

---**Q7:** Can you add a NOT NULL column to a table with existing data?

**A:** Only if you provide a DEFAULT value to fill the existing rows, otherwise it fails.

---**Q8:** What is the MERGE statement?

**A:** Standard SQL for Upsert. `MERGE INTO target USING source ON match... WHEN MATCHED UPDATE... WHEN NOT MATCHED INSERT...`.

---**Q9:** Difference between Primary Key and Unique Key?

**A:** Primary Key cannot be NULL and there is only one per table. Unique Key allows NULLs (usually one) and you can have multiple Unique constraints.

---**Q10:** What is a Temporary Table?

**A:** A table that exists only for the duration of a session or transaction (`CREATE TEMP TABLE`). Useful for intermediate processing steps.

# Chapter 7: Performance Tuning & Indexing ⚡
## 1. Detailed Module Content 📘
### Core Concepts
* **Indexes**: Special lookup tables that the search engine can use to speed up data retrieval. Like the index at the back of a book.
* **B-Tree Index**: The default. Good for ranges (`<`, `>`, `BETWEEN`) and equality.
* **Hash Index**: Good only for equality (`=`). Faster than B-Tree for exact lookups but useless for ranges.
* **Clustered Index**: Physically sorts the table data on disk. Only one per table (usually PK).
* **SARGable** (Search ARGument ABLE): Writing queries so they can use indexes (e.g., avoid `WHERE YEAR(date) = 2023`, use `WHERE date BETWEEN '2023-01-01' AND '2023-12-31'`).
* **EXPLAIN PLAN**: A command to see how the DB executes a query (Seq Scan vs Index Scan).

### Airline Realistic Example
Passengers search for flights by `Origin`, `Destination`, and `Date`. A **Composite Index** on `(Origin, Destination, Date)` would make this search instant. Without it, the DB scans the entire Flight schedule.

## 2. Top 10 Major Area (Theory) Questions 🧠
**Q1:** What is a 'Full Table Scan'?

**A:** When the database reads every single row in a table to find a match because no useful index exists. Slow on large tables.

---**Q2:** Why not index every column?

**A:** Indexes slow down write operations (INSERT/UPDATE/DELETE) because the index must be updated every time data changes. They also take up disk space.

---**Q3:** What is a Composite Index and does order matter?

**A:** An index on multiple columns (A, B). Order matters: the index helps queries filtering on A or (A and B), but not just B.

---**Q4:** Explain 'Index Selectivity'.

**A:** The ratio of unique values to total rows. High selectivity (e.g., Passport Number) is good for indexing. Low selectivity (e.g., Gender) is bad.

---**Q5:** What breaks SARGability?

**A:** Wrapping a column in a function (e.g., `UPPER(name) = 'TOM'`) prevents the index from being used.

---**Q6:** Difference between Clustered and Non-Clustered Index?

**A:** Clustered stores actual data rows at the leaf level (sorting the table). Non-clustered stores pointers to the data rows.

---**Q7:** What is a Covering Index?

**A:** An index that contains all the fields required by the query (in SELECT and WHERE), so the DB doesn't need to look up the main table at all.

---**Q8:** How does `EXPLAIN ANALYZE` help?

**A:** It runs the query and shows the actual time taken vs the estimated cost, helping identify bottlenecks.

---**Q9:** When should you rebuild indexes?

**A:** When they become fragmented (after many inserts/deletes), causing performance degradation.

---**Q10:** What is Query Partitioning elimination?

**A:** The optimizer's ability to skip scanning partitions that don't match the WHERE clause (in a partitioned table).

# Chapter 8: Database Architecture (OLAP/OLTP) 🏢
## 1. Detailed Module Content 📘
### Core Concepts
* **OLTP (Online Transactional Processing)**: Optimized for fast, frequent, small transactions (e.g., Booking a ticket). Highly Normalized (3NF).
* **OLAP (Online Analytical Processing)**: Optimized for heavy reading and aggregations (e.g., Revenue reports). Denormalized (Star Schema).
* **Normalization**: Splitting data to reduce redundancy.
  * **1NF**: Atomic values.
  * **2NF**: No partial dependencies.
  * **3NF**: No transitive dependencies.
* **Star Schema**: Central **Fact Table** (Metrics) connected to **Dimension Tables** (Attributes).
* **Snowflake Schema**: Like Star, but Dimensions are normalized (e.g., Airport Dim -> Country Dim).

### Airline Realistic Example
In an **OLTP** system, `Passenger` address might be in a separate table linked by ID. In an **OLAP** (Data Warehouse), we might keep `Passenger_City` inside the `Fact_Booking` table to avoid joining 5 tables just to report 'Revenue by City'.

## 2. Top 10 Major Area (Theory) Questions 🧠
**Q1:** Goal of Normalization vs Denormalization?

**A:** Normalization reduces redundancy and improves data integrity (Write-optimized). Denormalization reduces joins and improves read performance (Read-optimized).

---**Q2:** What is a Fact Table?

**A:** A table containing quantitative data (metrics) like 'Ticket Price', 'Duration', 'Delay Mins', and Foreign Keys to dimensions.

---**Q3:** What is a Dimension Table?

**A:** A table containing descriptive attributes like 'Aircraft Model', 'Airport Name', 'Passenger Gender'.

---**Q4:** Difference between Star and Snowflake schema?

**A:** Star: Dimensions are denormalized (flat). Simpler queries. Snowflake: Dimensions are normalized (branched). Saves space but requires more joins.

---**Q5:** What is a Surrogate Key?

**A:** An artificial PK (usually simple integer) used in DW instead of the business key (which might be complex or change).

---**Q6:** What is a Slowly Changing Dimension (SCD)?

**A:** How we handle data that changes over time (e.g., Aircraft assigned to a new hub).

---**Q7:** Explain SCD Type 1 vs Type 2.

**A:** Type 1 overwrites old data (No history). Type 2 creates a new row with dates (Full history).

---**Q8:** What is 3NF (Third Normal Form)?

**A:** A table is in 3NF if it is in 2NF and has no transitive dependencies (non-key columns depend only on the Key, not other non-key columns).

---**Q9:** Why are OLAP systems usually Column-oriented?

**A:** Because analytics queries usually aggregate specific columns (e.g., AVG(Price)) across millions of rows, rather than reading full rows.

---**Q10:** What is an ETL pipeline?

**A:** Extract, Transform, Load. The process of moving data from OLTP to OLAP.

# Chapter 9: Views & Materialized Views 🖼️
## 1. Detailed Module Content 📘
### Core Concepts
* **View**: A virtual table defined by a SQL query. It stores no data, just the logic. When queried, it runs the underlying query.
* **Materialized View (MV)**: A view where the result is physically stored on disk. It must be 'refreshed' to see new data. fast for heavy queries.
* **Security**: Views can hide sensitive columns (e.g., Credit Card numbers) from specific users.
* **Abstraction**: Views can simplify complex joins so analysts only query `SELECT * FROM simple_view`.

### Airline Realistic Example
Ground Crew needs a list of passengers for today's flights but shouldn't see ticket prices or payment info. We create a `View_Manifest` that selects only `Name`, `Seat`, and `Flight_No`.

## 2. Top 10 Major Area (Theory) Questions 🧠
**Q1:** Main difference between View and Materialized View?

**A:** Standard View calculates results on-the-fly (always current, slower). MV stores results (periodically updated, faster).

---**Q2:** Can you update data through a View?

**A:** Yes, if it's a 'Simple View' (1 table, no aggregates). 'Complex Views' (Joins, Aggregates) are generally read-only.

---**Q3:** Why use a View for Security?

**A:** You can grant permission on the View while denying permission on the underlying Table, effectively restricting row/column access.

---**Q4:** What happens to a View if you drop the underlying table?

**A:** The View becomes invalid (broken). Querying it will result in an error.

---**Q5:** What is `WITH CHECK OPTION` in a View?

**A:** It ensures that if you insert/update through the view, the new row must satisfy the View's WHERE clause.

---**Q6:** When should you refresh a Materialized View?

**A:** Depends on business needs: Real-time (Triggers), Scheduled (Cron job), or On-Demand.

---**Q7:** Can a View have an Index?

**A:** Standard Views cannot. Materialized Views CAN have indexes (because they are physical tables).

---**Q8:** What is a Recursive View?

**A:** A view that references itself (defined using CTE syntax), useful for hierarchies.

---**Q9:** Pros of Views?

**A:** Simplicity, Security, Consistency (centralized logic).

---**Q10:** Cons of Materialized Views?

**A:** Storage cost, Data latency (data is old until refreshed).

# Chapter 10: Transactions & Partitioning 📦
## 1. Detailed Module Content 📘
### Core Concepts
* **Transaction**: A logical unit of work. Must complete entirely or not at all.
* **ACID**:
  * **Atomicity**: All or Nothing.
  * **Consistency**: DB remains in valid state (constraints satisfied).
  * **Isolation**: Transactions don't interfere (Read Committed vs Serializable).
  * **Durability**: Saved data survives power loss.
* **Commands**: `BEGIN`, `COMMIT`, `ROLLBACK`, `SAVEPOINT`.
* **Partitioning**: Splitting a large table physically into smaller pieces (by Range, List, Hash) to improve performance.

### Airline Realistic Example
**Booking a Ticket**: 1. Deduct Money from User. 2. Reserve Seat on Plane. 3. Generate Ticket. If Step 1 fails, nothing happens. If Step 3 fails, Step 1 (Money) must be refunded (Rolled back). The seat cannot be booked by two people at once (Isolation).

## 2. Top 10 Major Area (Theory) Questions 🧠
**Q1:** What happens if you don't COMMIT a transaction?

**A:** The changes remain temporary and visible only to your session. If the session ends/crashes, they are Rolled Back (lost).

---**Q2:** What is a Deadlock?

**A:** Process A waits for Resource B (held by Process B), while Process B waits for Resource A (held by Process A). Both are stuck forever.

---**Q3:** Difference between Horizontal and Vertical Partitioning?

**A:** Horizontal splits rows (e.g., Years 2020, 2021). Vertical splits columns (e.g., Frequent access cols vs Rare access cols).

---**Q4:** What is Sharding?

**A:** Horizontal partitioning across *different servers* (Distributed database).

---**Q5:** Explain 'Dirty Read'.

**A:** Reading data that another transaction has written but not yet Committed. (Risk: The other transaction might Rollback).

---**Q6:** What is the 'Serializable' isolation level?

**A:** The strictest level. It emulates serial execution (one after another), preventing all concurrency anomalies but hurting performance.

---**Q7:** Why partition by Date for Log tables?

**A:** Allows 'Partition Pruning' (Querying Jan 2023 only scans the Jan partition) and easy archiving (Drop Partition Jan 2020).

---**Q8:** What is a Savepoint?

**A:** A marker inside a transaction. You can rollback to the savepoint without rolling back the entire transaction.

---**Q9:** What is Optimistic Concurrency Control?

**A:** Assumes conflicts are rare. Checks version/timestamp at the end (Commit time) rather than locking rows at the start.

---**Q10:** Can you partition an existing non-partitioned table?

**A:** Usually no. You typically create a new partitioned table and migrate data.
