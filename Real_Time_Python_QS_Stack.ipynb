## Question 27: Look-and-Say Sequence

### Lecture Note
The Look-and-Say sequence describes the previous term.

Example:

11 → "two 1s" → 21  

**Key Learning Areas**
- String grouping
- Iterative sequence generation

### Python Solution
def look_and_say(n):

    result = "1"

    for _ in range(n - 1):

        current = ""
        count = 1

        for i in range(1, len(result)):

            if result[i] == result[i - 1]:
                count += 1
            else:
                current += str(count) + result[i - 1]
                count = 1

        current += str(count) + result[-1]
        result = current

    return result

        left += 1

- Move only to adjacent numbers in the next row.

**Key Learning Areas**
- Dynamic programming
- Bottom-up optimization

def min_path_sum(triangle):

    for i in range(len(triangle)-2, -1, -1):

        for j in range(len(triangle[i])):

            triangle[i][j] += min(
                triangle[i+1][j],
                triangle[i+1][j+1]

from collections import OrderedDict

class LRUCache:

    def __init__(self, capacity):

        self.cache = OrderedDict()
        self.capacity = capacity

    def get(self, key):

        if key not in self.cache:
            return -1

        self.cache.move_to_end(key)
