# 🗄️ SQL Interview Study Guide: Top 10 Topics for Data Engineers

This guide follows the same structure as your Python study plan, focusing on the essential SQL pillars required for building and maintaining data pipelines.

---

## 1. Joins and Set Operations 🧩
* **Join Types:** Understand `INNER`, `LEFT`, `RIGHT`, `FULL OUTER`, and `CROSS JOIN` (Cartesian products).
* **Self-Joins:** Useful for hierarchical data or comparing rows in the same table.
* **Set Operators:** Using `UNION`, `UNION ALL`, `INTERSECT`, and `EXCEPT`.

```python
-- Example: Finding users who placed orders (Inner Join)
SELECT users.name, orders.order_date
FROM users
INNER JOIN orders ON users.id = orders.user_id;
```

## 2. Aggregations and Grouping 📊
* **Functions:** `COUNT()`, `SUM()`, `AVG()`, `MIN()`, `MAX()`.
* **Filtering:** The difference between `WHERE` (pre-aggregation) and `HAVING` (post-aggregation).

```python
-- Example: Total revenue per product category with more than $1000 sales
SELECT category, SUM(price) as total_revenue
FROM sales
WHERE status = 'completed'
GROUP BY category
HAVING SUM(price) > 1000;
```

## 3. Window Functions (Analytical Functions) 🪟
* **Ranking:** `ROW_NUMBER()`, `RANK()`, `DENSE_RANK()`.
* **Positional:** `LEAD()` and `LAG()` for time-series comparisons.
* **Aggregates over Partition:** `SUM(...) OVER(PARTITION BY ...)`.

```python
-- Example: Ranking employees by salary within their department
SELECT name, dept, salary, 
       RANK() OVER(PARTITION BY dept ORDER BY salary DESC) as salary_rank
FROM employees;
```

## 4. Subqueries and CTEs 🏗️
* **CTEs (WITH clause):** Preferred for readability and modularity in Data Engineering pipelines.
* **Correlated Subqueries:** Understanding performance impacts.

```python
-- Example: Using a CTE for cleaner logic
WITH HighValueOrders AS (
    SELECT user_id, amount
    FROM orders
    WHERE amount > 500
)
SELECT u.name, hvo.amount
FROM users u
JOIN HighValueOrders hvo ON u.id = hvo.user_id;
```

## 5. Data Types and Casting 🔢
* **Handling Grains:** `DATE`, `TIMESTAMP`, `INTERVAL`.
* **Type Conversion:** `CAST()` and database-specific shorthand (e.g., `::int`).

## 6. DDL and DML Operations 🛠️
* **Schema Management:** `CREATE`, `ALTER`, `DROP`.
* **Data Modification:** `INSERT`, `UPDATE`, `DELETE`.
* **Upsert Logic:** Handling conflicts using `ON CONFLICT` or `MERGE`.

## 7. Performance Tuning and Indexing ⚡
* **Indexes:** B-Tree, Hash, and GIN indexes.
* **Analysis:** Reading `EXPLAIN ANALYZE` outputs to find bottlenecks.

## 8. Database Architecture for DE 🏢
* **Schema Design:** Star Schema (Fact and Dimension tables) vs. Snowflake Schema.
* **Normalization:** 1NF, 2NF, 3NF concepts for transactional systems.

## 9. Views and Materialized Views 🖼️
* **Views:** Virtual layers for security and abstraction.
* **Materialized Views:** Physical storage of query results for performance.

## 10. Transactions and Partitioning 📦
* **ACID:** Atomicity, Consistency, Isolation, Durability.
* **Table Partitioning:** Splitting tables by date or region to speed up big data queries.
