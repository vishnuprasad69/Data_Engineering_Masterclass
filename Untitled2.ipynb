# Data Engineering SQL Interview Handbook

## Overview
This document provides a structured set of SQL problems commonly asked in Data Engineering interviews, along with optimized solutions and key engineering insights.

---

## 1. Histogram of User Activity
**Goal:** Count how many users fall into each activity bucket.

```sql
WITH activity_counts AS (
  SELECT user_id, COUNT(*) AS activity_count
  FROM activity_table
  WHERE activity_date >= '2022-01-01' AND activity_date < '2023-01-01'
  GROUP BY user_id
)
SELECT activity_count, COUNT(user_id) AS user_count
FROM activity_counts
GROUP BY activity_count
ORDER BY activity_count;
```

**Insight:** Partition large datasets by date to improve performance and reduce scan size.

---

## 2. Multi-Skill Matching
**Goal:** Identify users who meet multiple criteria.

```sql
SELECT user_id
FROM skills_table
WHERE skill IN ('Skill1', 'Skill2', 'Skill3')
GROUP BY user_id
HAVING COUNT(DISTINCT skill) = 3;
```

**Insight:** Use DISTINCT to avoid duplicate data issues.

---

## 3. Finding Missing Relationships
**Goal:** Identify records that do not have matching entries.

```sql
SELECT a.id
FROM table_a a
LEFT JOIN table_b b ON a.id = b.id
WHERE b.id IS NULL;
```

**Insight:** Prefer LEFT JOIN over NOT IN to avoid NULL-related issues.

---

## 4. Identifying Incomplete Records
```sql
SELECT item_id
FROM items
WHERE completion_date IS NULL;
```

**Insight:** Columnar storage formats optimize NULL checks using metadata.

---

## 5. Conditional Aggregation
```sql
SELECT 
  SUM(CASE WHEN category = 'A' THEN 1 ELSE 0 END) AS category_a,
  SUM(CASE WHEN category IN ('B','C') THEN 1 ELSE 0 END) AS category_bc
FROM dataset;
```

**Insight:** Enables single-pass aggregation.

---

## 6. Date Range Analysis
```sql
SELECT 
  user_id,
  MAX(date_col) - MIN(date_col) AS duration
FROM dataset
WHERE date_col >= '2021-01-01' AND date_col < '2022-01-01'
GROUP BY user_id
HAVING COUNT(*) >= 2;
```

**Insight:** Avoid functions in WHERE clauses for better index usage.

---

## 7. Top-N Queries
```sql
SELECT user_id, COUNT(*) AS cnt
FROM dataset
WHERE date_col BETWEEN '2022-08-01' AND '2022-08-31'
GROUP BY user_id
ORDER BY cnt DESC
LIMIT 2;
```

**Insight:** Engines optimize small LIMIT queries using Top-K strategies.

---

## 8. Duplicate Detection
```sql
SELECT col1
FROM dataset
GROUP BY col1, col2, col3
HAVING COUNT(*) > 1;
```

**Insight:** Hash large text fields to improve performance.

---

## 9. Fact-Dimension Join
```sql
SELECT d.dimension, COUNT(f.id)
FROM fact_table f
JOIN dimension_table d ON f.key = d.key
GROUP BY d.dimension;
```

**Insight:** Use broadcast joins for small dimension tables.

---

## 10. Time-Based Aggregation
```sql
SELECT 
  EXTRACT(MONTH FROM date_col) AS month,
  id,
  ROUND(AVG(metric), 2)
FROM dataset
GROUP BY month, id;
```

**Insight:** Precompute aggregations for dashboards.

---

## 11. Self Join Comparison
```sql
SELECT a.name
FROM employees a
JOIN employees b ON a.manager_id = b.id
WHERE a.salary > b.salary;
```

**Insight:** Flatten hierarchies for large-scale systems.

---

## 12. Conditional Summation
```sql
SELECT id,
  SUM(CASE WHEN type='credit' THEN amount ELSE -amount END)
FROM transactions
GROUP BY id;
```

**Insight:** Ensure idempotent processing in pipelines.

---

## 13. Ratio Calculation
```sql
SELECT 
  id,
  SUM(clicks) / NULLIF(SUM(impressions),0) AS ratio
FROM dataset
GROUP BY id;
```

**Insight:** Always guard against division by zero.

---

## 14. Date Interval Logic
```sql
SELECT id
FROM table_a a
JOIN table_b b ON a.key = b.key
WHERE b.date_col = a.date_col + INTERVAL '1 day';
```

**Insight:** Standardize timestamps (UTC) in distributed systems.

---

## 15. Left Join Aggregation
```sql
SELECT a.id, COUNT(b.id)
FROM table_a a
LEFT JOIN table_b b ON a.id = b.id
GROUP BY a.id;
```

**Insight:** Apply filters in JOIN condition to preserve NULL rows.

---

## 16. Min-Max Difference
```sql
SELECT category,
  MAX(value) - MIN(value)
FROM dataset
GROUP BY category;
```

**Insight:** Columnar formats optimize MIN/MAX using metadata.

---

## 17. Weighted Average
```sql
SELECT 
  SUM(value * weight) / SUM(weight)
FROM dataset;
```

**Insight:** Always cast to decimal to avoid integer truncation.

---

## 18. Row-Level Computation
```sql
SELECT col1, col2 - col3 AS result
FROM dataset;
```

**Insight:** Efficient in distributed engines via vectorized execution.

---

## 19. Filtered Aggregation
```sql
SELECT category, SUM(metric)
FROM dataset
WHERE metric_condition
GROUP BY category;
```

**Insight:** Apply filters early to reduce shuffle size.

---

## 20. Data Formatting
```sql
SELECT CONCAT('$', ROUND(SUM(value)/1000000), ' million')
FROM dataset;
```

**Insight:** Keep formatting in presentation layer, not data layer.

---

## 21. Multi-Level Aggregation
```sql
WITH temp AS (
  SELECT id, COUNT(*) AS cnt
  FROM dataset
  GROUP BY id
)
SELECT COUNT(*)
FROM temp
WHERE cnt >= 3;
```

**Insight:** Be cautious of global aggregations causing bottlenecks.

---

## Final Notes
- Focus on **query efficiency + scalability**
- Think in terms of **distributed systems (Spark, Glue, Databricks)**
- Always consider **data size, skew, and execution plan**
