# Big Data Fundamentals: HDFS, ACID, Batch vs Streaming, and Schema Validation

_Generated on: 2026-01-28_


## 1️⃣ HDFS (Hadoop Distributed File System)

### Definition
HDFS is a distributed file system designed to store and process large volumes of data across multiple machines with fault tolerance.

### Key Points
- Stores data in large blocks (128/256 MB)
- Data is replicated across nodes (default = 3)
- Optimized for high-throughput, sequential access
- Designed for **write once, read many**

### Core Components
- **NameNode** → Metadata management
- **DataNode** → Actual data storage
- **Standby/Secondary NameNode** → Metadata checkpointing

### Limitations
- Not suitable for small files
- Not ACID compliant
- High latency for random reads


## 2️⃣ ACID Properties

### Definition
ACID defines the guarantees that ensure reliable and consistent database transactions.

### Breakdown
- **Atomicity** → All or nothing
- **Consistency** → Data rules are always maintained
- **Isolation** → Concurrent transactions don’t interfere
- **Durability** → Committed data survives failures

### Big Data Context
- **Traditional RDBMS** → Fully ACID
- **HDFS** → Not ACID
- **Delta Lake / Iceberg / Hudi** → ACID on data lakes


## 3️⃣ Batch Processing

### Definition
Batch processing handles large volumes of data together at scheduled intervals.

### Characteristics
- Processes historical data
- Runs on a schedule
- High throughput, higher latency
- Simpler and stable

### Examples
- Daily ETL jobs
- Report generation
- Data warehouse loads

### Technologies
- Spark (batch)
- Hive
- Hadoop MapReduce
- AWS Glue
- Databricks Jobs


## 4️⃣ Streaming Processing

### Definition
Streaming processing handles data continuously in real time as it arrives.

### Characteristics
- Processes unbounded data
- Low latency
- Event-driven or micro-batch
- Requires state management

### Examples
- Real-time dashboards
- Fraud detection
- Live notifications

### Technologies
- Kafka
- Spark Structured Streaming
- Flink
- Kinesis / Event Hubs


## 5️⃣ Batch vs Streaming (Interview Table)

| **Aspect** | **Batch** | **Streaming** |
|-----------|-----------|---------------|
| **Data type** | Bounded | Unbounded |
| **Latency** | Minutes–hours | Milliseconds–seconds |
| **Execution** | Scheduled | Continuous |
| **Complexity** | Low | High |
| **Use case** | Reports, ETL | Alerts, real-time analytics |


## 6️⃣ Schema Validation

### Definition
Schema validation ensures incoming data strictly follows the expected structure, data types, and constraints.

### What Is Validated
- Column names
- Data types
- Nullability
- Missing / extra fields
- Format correctness

### Why It Matters
- Prevents bad data ingestion
- Improves data quality
- Protects downstream pipelines
- Supports **Consistency (C in ACID)**

### Implementation Examples
- Spark `StructType`
- Glue schema enforcement
- Delta Lake schema enforcement
- Avro / Protobuf with Schema Registry
